## SKU навигатор

 ABC/XYZ-анализ и поддержка решений в Power BI

 Аннотация

«Навигатор ассортимента» — аналитический инструмент, который помогает интернет-магазину принимать обоснованные решения о составе ассортимента. Проект использует транзакционные данные **Online Retail II** и объединяет подготовку данных, ABC/XYZ-анализ и интерактивную визуализацию в Power BI.

**Актуальность проекта** связана с необходимостью управлять растущим количеством товарных позиций. Редкие продажи и низкая выручка не всегда означают, что товар следует вывести: он может быть новым, сезонным или важным дополнением к другим покупкам.

**Цель проекта** — определить, какие товары следует приоритетно поддерживать в ассортименте, какие требуют осторожного управления, а какие можно рассматривать как кандидатов на сокращение.

**Основные задачи:**
- разработать воспроизводимый ETL-процесс очистки и подготовки данных;
- оценить структуру продаж и распределить товары по экономической значимости и вариативности продаж;
- сформировать объяснимые критерии отбора кандидатов с учетом дополнительных факторов риска;
- создать инструмент Power BI для самостоятельного выбора критериев и сравнения сценариев заказчиком;
- оценить возможную долю сокращения SKU и связанную с ними историческую выручку.

**Ожидаемый результат** — аналитический отчет и интерактивная BI-система, позволяющая исследовать ассортимент, проверять отдельные товары и формировать список кандидатов на сокращение. Сокращение примерно 20% SKU рассматривается как ориентир, а не обязательный результат. Историческая выручка кандидатов не интерпретируется как гарантированная будущая потеря, а рекомендации требуют проверки с учетом остатков, маржинальности и условий поставок.

### Дубликаты

Полные повторы определяем по восьми бизнес-полям. Отдельно считаем межлистовые копии — строки, которые встречаются в обоих листах выгрузки.

```python
cols = ["Invoice", "StockCode", "Description", "Quantity",
        "InvoiceDate", "Price", "Customer ID", "Country"]

dup_full  = raw.duplicated(cols)                     # сверх первого
dup_any   = raw.duplicated(cols, keep=False)         # все в группах
dup_cross = raw.duplicated(cols + ["source_sheet"])  # внутрилистовые

base     = raw.loc[~dup_cross]
dup_base = base.duplicated(cols)                     # межлистовые

n_full  = int(dup_full.sum())
n_intra = int(dup_cross.sum())                       # ← было "Межлистовых"
n_inter = int(dup_base.sum())                        # ← было "после исключения"

display(pd.Series({
    "Полных повторов сверх первого":       n_full,
    "Внутрилистовых копий":                 n_intra,
    "Доля внутрилистовых, %":               round(100 * n_intra / n_full, 1),
    "Межлистовых копий":                    n_inter,
    "Доля межлистовых, %":                  round(100 * n_inter / n_full, 1),
}, name="Количество").to_frame())

display(
    raw.loc[dup_any, "InvoiceDate"]
       .dt.date.value_counts().head(5)
       .rename("Строк в группах полных повторов").to_frame()
)
```

Найдено 34 335 полных повторов сверх первого. Из них 12 133 (35,3%) — внутрилистовые копии (строка дважды в одном листе) и 22 202 (64,7%) — межлистовые. Межлистовые сконцентрированы в начале декабря 2010 и отражают перекрытие двух выгрузок по декабрю.

Связь с отменами и влияние оставшихся повторов

```python
# Нужные маски повторов внутри base
dup_in_base_any   = base.duplicated(cols, keep=False)
dup_in_base_first = base.duplicated(cols)

# Производные на base — считаем один раз
is_cancel = (
    base["Invoice"].str.strip().str[:1].str.upper()
        .eq("C").fillna(False).astype(bool)
)
amount  = base["Quantity"].astype("float64") * base["Price"]
abs_qty = base["Quantity"].abs()

sales_mask = ~is_cancel & base["Quantity"].gt(0) & base["Price"].gt(0)
canc_mask  = is_cancel & base["Quantity"].lt(0) & base["Price"].gt(0)

repeated = base.loc[sales_mask & dup_in_base_any]
cancels  = base.loc[canc_mask]

match = ["Customer ID", "StockCode", "Price", "AbsQty"]

# Первая повторяющаяся продажа по (клиент, SKU, цена, |Qty|)
rep_sub = (
    repeated.loc[repeated["Customer ID"].notna(),
                 ["Customer ID", "StockCode", "Price", "InvoiceDate"]]
    .copy()
)
rep_sub["AbsQty"] = abs_qty.loc[rep_sub.index].to_numpy()

first_sale = (
    rep_sub.groupby(match, observed=True)["InvoiceDate"]
    .min().rename("FirstSale").reset_index()
)

canc_sub = (
    cancels.loc[cancels["Customer ID"].notna(),
                ["Invoice", "Customer ID", "StockCode", "Price", "InvoiceDate"]]
    .copy()
)
canc_sub["AbsQty"] = abs_qty.loc[canc_sub.index].to_numpy()

candidates = (
    canc_sub
    .merge(first_sale, on=match, how="inner", validate="many_to_one")
    .loc[lambda x: x["InvoiceDate"] >= x["FirstSale"]]
)

display(pd.Series({
    "Отрицательных строк C-счетов": len(cancels),
    "Из них без клиента — вне сопоставления":
        int(cancels["Customer ID"].isna().sum()),
    "Строк отмен с возможной связью": len(candidates),
    "Затронуто счетов отмен": candidates["Invoice"].nunique(),
}, name="Количество").to_frame())


# ─────────────────────────────────────────────────────
# Влияние удаления оставшихся повторов
# Исправление: отрицательные суммы считаем через маску, а не clip.
# clip(upper=0) в некоторых версиях pandas не суммирует negatives корректно.
# ─────────────────────────────────────────────────────

def metrics_from(amount_series, rows):
    return pd.Series({
        "Строк": rows,
        "Положительные суммы":
            amount_series[amount_series > 0].sum(),
        "Отрицательные суммы, модуль":
            abs(amount_series[amount_series < 0].sum()),
        "Сальдо операций":
            amount_series.sum(),
    })

keep = ~dup_in_base_first

impact = pd.concat(
    [
        metrics_from(amount, len(amount)),
        metrics_from(amount.loc[keep], int(keep.sum())),
    ],
    axis=1,
    keys=["С повторами", "Без полных повторов"],
)
impact["Изменение"] = impact.iloc[:, 1] - impact.iloc[:, 0]
impact["Изменение, %"] = (
    impact["Изменение"]
    / impact["С повторами"].abs().replace(0, float("nan")) * 100
)
display(impact.round(2))
```

22 202 межлистовые копии и 12 133 внутрилистовых дубля прошли проверку: отмены почти не связаны с задвоенными продажами (только 0,8% потенциально совпадают), а влияние на суммы не превышает 3,7% по любому показателю. Поэтому принимаем решение удалить все полные совпадения, оставив один экземпляр каждой записи.

```python
# Финальная очистка: один экземпляр по восьми полям
remove = raw.duplicated(subset=cols, keep="first")

removed_duplicates = raw.loc[remove].copy()
clean = raw.loc[~remove].copy().reset_index(drop=True)

assert not clean.duplicated(subset=cols).any()
assert len(raw) == len(clean) + len(removed_duplicates)

display(pd.Series({
    "Исходных строк": len(raw),
    "Удалено полных повторов": len(removed_duplicates),
    "Осталось строк": len(clean),
    "Полных повторов после очистки": int(clean.duplicated(cols).sum()),
}, name="Количество").to_frame())
```

Удалено 34 335 полных повторов, осталось 1 033 036 строк. Из них 22 202 — межлистовые копии (пересечение выгрузок), 12 133 — внутрилистовые повторы. Проверка показала: отмены почти не связаны с задвоенными продажами (0,8% потенциальных пар), а влияние на суммы не превышает 3,7% по любому показателю. Поэтому принимаем решение удалить все полные совпадения, оставив один экземпляр. Две пары строк с разными описаниями сохранены — они не являются полными дублями. Правило применено к продажам и отменам одинаково. Исходные и исключённые записи сохранены для проверки. Далее используем clean.

### Отмены и отрицательные Quantity

Проверка количеств и цен

```python
cols = [
    "Invoice", "StockCode", "Description", "Quantity",
    "InvoiceDate", "Price", "Customer ID", "Country",
]

clean["IsCancel"] = (
    clean["Invoice"].astype("string").str.strip().str.upper()
         .str.startswith("C", na=False)
)

display(pd.crosstab(
    clean["IsCancel"], clean["Quantity"].lt(0),
    rownames=["C-счет"], colnames=["Quantity < 0"],
))
print("Нулевых Quantity:", clean["Quantity"].eq(0).sum())

# Проверяем исключения.
negative_other = clean.loc[
    ~clean["IsCancel"] & clean["Quantity"].lt(0)
]
display(pd.Series({
    "Отрицательных строк без C": len(negative_other),
    "Из них без клиента": negative_other["Customer ID"].isna().sum(),
    "Из них с нулевой ценой": negative_other["Price"].eq(0).sum(),
}, name="Строк").to_frame())

display(
    negative_other["Description"].value_counts(dropna=False)
        .head(8).rename("Строк").to_frame()
)
display(clean.loc[
    clean["IsCancel"] & clean["Quantity"].ge(0), cols
])

# Цены среди положительных операций без C.
sales_base = clean.loc[
    ~clean["IsCancel"] & clean["Quantity"].gt(0)
].copy()
price = sales_base["Price"]

display(pd.Series({
    "Положительная цена": price.gt(0).sum(),
    "Нулевая цена": price.eq(0).sum(),
    "Отрицательная цена": price.lt(0).sum(),
    "Цена отсутствует": price.isna().sum(),
    "Нулевая цена, клиент известен":
        (price.eq(0) & sales_base["Customer ID"].notna()).sum(),
}, name="Строк").to_frame())

display(sales_base.loc[price.lt(0) | price.isna(), cols])
display(
    sales_base.loc[price.eq(0), "Description"]
        .value_counts(dropna=False).head(8).rename("Строк").to_frame()
)
```

Единственный положительный C-счет — ручная операция на 373,57, не идентифицированная как товарная покупка.

Проверка нетоварных позиций

Итоговая обработка и контроль выборок

```python
rows_before = len(clean)

clean["IsCancel"] = (
    clean["Invoice"].astype("string").str.strip().str.upper()
         .str.startswith("C", na=False)
)

stock = clean["StockCode"].astype("string").str.strip().str.upper()

exclude_codes = {
    "M", "DOT", "POST", "AMAZONFEE", "C2", "B",
    "ADJUST", "ADJUST2", "BANK CHARGES", "D",
    "S", "TEST001", "TEST002",
}

nonproduct = (
    stock.isin(exclude_codes)
    | stock.str.startswith("GIFT_", na=False)
)

# Допустимые сочетания: продажа без C или отрицательная операция с C.
valid_operation = (
    (~clean["IsCancel"] & clean["Quantity"].gt(0))
    | (clean["IsCancel"] & clean["Quantity"].lt(0))
)

keep = valid_operation & clean["Price"].gt(0) & ~nonproduct

clean = clean.loc[keep].copy().reset_index(drop=True)
clean["Amount"] = clean["Quantity"] * clean["Price"]

# Нормализация StockCode — как в ETL (clean.py).
# Без неё 85123a и 85123A считаются разными SKU:
# в ноутбуке получается на 170 SKU больше, чем в parquet/PBI.
clean["StockCode"] = (
    clean["StockCode"].astype("string").str.strip().str.upper()
)

# Контроль правил очистки.
assert clean["Price"].gt(0).all()
assert clean.loc[~clean["IsCancel"], "Quantity"].gt(0).all()
assert clean.loc[clean["IsCancel"], "Quantity"].lt(0).all()

display(pd.Series({
    "Строк до обработки": rows_before,
    "Исключено строк": rows_before - len(clean),
    "Осталось строк": len(clean),
    "Товарные продажи": (~clean["IsCancel"]).sum(),
    "Товарные отмены / возвраты": clean["IsCancel"].sum(),
    "Продажи с известным клиентом":
        (~clean["IsCancel"] & clean["Customer ID"].notna()).sum(),
    "Продажи без клиента":
        (~clean["IsCancel"] & clean["Customer ID"].isna()).sum(),
}, name="Количество").to_frame())

display(
    clean.groupby("IsCancel").agg(
        rows=("Invoice", "size"),
        amount=("Amount", "sum"),
    ).rename(index={False: "Продажи", True: "Отмены / возвраты"})
)
display(clean.head())
```

## Исследовательский анализ продаж

### Основные показатели за два годовых периода

Начнём с масштаба бизнеса и сравнения двух лет: сколько продали, сколько получили выручки, сколько было заказов, товаров и известных покупателей.

Правила расчёта:

выручку, заказы и проданные единицы считаем по продажам без отмен;

возвраты показываем отдельно, положительной суммой;

средний чек — выручка продаж / число заказов;

количество покупателей — только по известным Customer ID.

```python
sales = clean.loc[~clean["IsCancel"]]
returns = clean.loc[clean["IsCancel"]]

# Основные показатели положительных товарных продаж.
summary = (
    sales.groupby("Period", observed=True)
    .agg(
        revenue=("Amount", "sum"),
        units_sold=("Quantity", "sum"),
        orders=("Invoice", "nunique"),
        sku_sold=("StockCode", "nunique"),
        known_customers=("Customer ID", "nunique"),
        sales_rows=("Invoice", "size"),
        rows_with_customer=("Customer ID", "count"),
    )
)

# Отмены / возвраты, зарегистрированные в каждом периоде.
return_amount = (
    returns.groupby("Period", observed=True)["Amount"]
    .sum()
    .abs()
)

summary["return_amount"] = return_amount.reindex(
    summary.index, fill_value=0
)

summary["net_amount"] = (
    summary["revenue"] - summary["return_amount"]
)

summary["avg_order_value"] = (
    summary["revenue"] / summary["orders"]
)

summary["units_per_order"] = (
    summary["units_sold"] / summary["orders"]
)

summary["customer_id_coverage_pct"] = (
    100 * summary["rows_with_customer"] / summary["sales_rows"]
)

# Удобный формат: показатели по строкам, периоды по столбцам.
labels = {
    "revenue": "Выручка продаж",
    "return_amount": "Сумма отмен / возвратов",
    "net_amount": "Сальдо товарных операций",
    "units_sold": "Продано единиц",
    "orders": "Заказов с товарными продажами",
    "sku_sold": "SKU с продажами",
    "known_customers": "Известных покупателей",
    "avg_order_value": "Средний товарный чек",
    "units_per_order": "Единиц на заказ",
    "customer_id_coverage_pct": "Строк продаж с Customer ID, %",
}

report = summary[list(labels)].rename(columns=labels).T

display(report.round(2))
```

#### Выводы по основным показателям

- **Выручка выросла на 2,5%**, до 9,63 млн, при сокращении числа заказов на **4,0%**. Рост поддержан увеличением среднего товарного чека на **6,8%**.
- **Продано на 6,7% меньше единиц**, а единиц на заказ стало меньше на **2,8%**. Рост выручки не связан с ростом физического объёма; влияние цен и структуры продаж предстоит проверить.
- **Количество SKU с продажами сократилось на 6,8%** — с 4 073 до 3 797. Выручка при этом выросла, но это ещё не доказывает, что сужение ассортимента улучшило результат.
- **Сумма отмен/возвратов выросла на 25,5%**: отношение к выручке увеличилось с **2,56% до 3,13%**. Поэтому сальдо выросло слабее выручки — на **1,9%**.
- Известных покупателей стало на **1,3% больше**, однако покрытие строк продаж клиентскими ID снизилось с **80,16% до 75,17%**. Динамику клиентской базы нужно интерпретировать осторожно.

### Структура продаж

проверим:

товары: лидеры по выручке и количеству, концентрация выручки в топ-10/50/100 SKU;
географию: вклад стран в выручку;
отмены/возвраты: их соотношение с продажами и товары с наибольшими суммами возвратов.
Все показатели рассчитываем отдельно по двум годовым периодам. clean не изменяем, ABC-классы пока не присваиваем.

```python
# Фильтры продаж и возвратов по знаку количества
sale = clean["Quantity"] > 0
ret = clean["Quantity"] < 0

# Самое частое непустое название SKU; при равенстве — первое по алфавиту
names = (
    clean.assign(ProductName=clean["Description"].astype("string")
                 .str.replace(r"\s+", " ", regex=True).str.strip()
                 .replace("", pd.NA))
    .dropna(subset=["ProductName"])
    .groupby(["Period", "StockCode"], observed=True)["ProductName"]
    .agg(lambda x: x.mode().iloc[0])
)


# Общий расчёт продаж по товарам или странам
def sales_structure(column):
    table = (
        clean.loc[sale]
        .groupby(["Period", column], observed=True, dropna=False)
        .agg(revenue=("Amount", "sum"),
             units=("Quantity", "sum"),
             orders=("Invoice", "nunique"))
    )
    total = table.groupby(level="Period", observed=True)["revenue"].transform("sum")
    table["revenue_share_pct"] = 100 * table["revenue"] / total.where(total.ne(0))
    return table

# 1. Продажи по товарам и концентрация выручки
sku_structure = sales_structure("StockCode").join(names)
sku_structure["ProductName"] = sku_structure["ProductName"].fillna("Без описания")
sku_structure = sku_structure[
    ["ProductName", "revenue", "units", "orders", "revenue_share_pct"]
]

concentration = []
for period, group in sku_structure.groupby(level="Period", observed=True):
    ranked = group.sort_values("revenue", ascending=False)
    concentration.append({
        "Period": period,
        "sku_sold": len(ranked),
        **{f"top{n}_share_pct": ranked.head(n)["revenue_share_pct"].sum()
           for n in [10, 50, 100]}
    })

concentration = pd.DataFrame(concentration).set_index("Period")
print("КОНЦЕНТРАЦИЯ ВЫРУЧКИ ПО SKU — ДО ВОЗВРАТОВ")
display(concentration.round(2))

# 2. Продажи по странам и средний чек до возвратов
country_structure = sales_structure("Country")
country_structure["avg_order_value"] = (
    country_structure["revenue"] / country_structure["orders"]
)

# 3. Суммы продаж и возвратов по периодам
returns_summary = pd.DataFrame({
    "sales_revenue": clean.loc[sale].groupby("Period", observed=True)["Amount"].sum(),
    "return_amount": clean.loc[ret].groupby("Period", observed=True)["Amount"].sum().abs()
}).fillna(0)

returns_summary["returns_to_sales_pct"] = (
    100 * returns_summary["return_amount"]
    / returns_summary["sales_revenue"].where(returns_summary["sales_revenue"].gt(0))
)

print("\nВОЗВРАТЫ ОТНОСИТЕЛЬНО ПРОДАЖ")
display(returns_summary.round(2))

# 4. Возвраты по SKU, включая товары без продаж в периоде
sku_returns = (
    clean.loc[ret].groupby(["Period", "StockCode"], observed=True)["Amount"]
    .sum().abs().rename("return_amount")
)

sku_operations = (
    sku_structure[["revenue"]].join(sku_returns, how="outer").fillna(0)
)
sku_operations["net_amount"] = (
    sku_operations["revenue"] - sku_operations["return_amount"]
)
sku_operations["returns_to_sales_pct"] = (
    100 * sku_operations["return_amount"]
    / sku_operations["revenue"].where(sku_operations["revenue"].gt(0))
)
sku_operations = sku_operations.join(names)
sku_operations["ProductName"] = sku_operations["ProductName"].fillna("Без описания")
sku_operations = sku_operations[
    ["ProductName", "revenue", "return_amount", "net_amount", "returns_to_sales_pct"]
]

# Выводим все рейтинги одним циклом
reports = [
    ("SKU ПО ВЫРУЧКЕ ДО ВОЗВРАТОВ", sku_structure, "revenue"),
    ("SKU ПО ПРОДАННЫМ ЕДИНИЦАМ ДО ВОЗВРАТОВ", sku_structure, "units"),
    ("СТРАН ПО ВЫРУЧКЕ ДО ВОЗВРАТОВ", country_structure, "revenue"),
    ("SKU ПО СУММЕ ВОЗВРАТОВ", sku_operations, "return_amount"),
]

with pd.option_context("display.max_colwidth", 80, "display.max_columns", None):
    for title, table, metric in reports:
        for period, group in table.groupby(level="Period", observed=True):
            print(f"\n{period}: ТОП-10 {title}")
            display(group.droplevel("Period").nlargest(10, metric).round(2))
```

### Проверка крупных операций и их влияния

```python
# Крупнейшие товарные строки, а не заказы целиком
cols = ["Invoice", "InvoiceDate", "StockCode", "Quantity", "Amount"]

# Отдельно считаем суммы продаж и возвратов
for title, mask in [
    ("ПРОДАЖИ", clean["Amount"] > 0),
    ("ВОЗВРАТЫ", clean["Amount"] < 0),
]:
    amounts = clean.loc[mask, "Amount"].abs()
    total = amounts.sum()
    top_sum = amounts.nlargest(5).sum()

    # Сортировка по модулю: самые крупные операции первыми
    top = (
        clean.loc[mask, cols]
        .sort_values("Amount", key=lambda x: x.abs(), ascending=False)
        .head(5)
    )

    print(f"\n{title}: общая сумма — {total:,.2f} GBP")
    print(f"5 крупнейших строк — {top_sum:,.2f} GBP")
    print(f"Их доля в общей сумме — {top_sum / total:.1%}" if total else "Операций нет")
    display(top)

# Итоговый денежный результат с учётом всех возвратов
print(f"\nВыручка после возвратов: {clean['Amount'].sum():,.2f} GBP")
```

## ABC-анализ ассортимента

Используем готовую sku_structure: в ней уже рассчитана выручка каждого SKU отдельно за каждый годовой период, до возвратов.

Сортируем товары по убыванию выручки и присваиваем классы:

A — накопленная доля до 80%;
B — свыше 80% до 95%;
C — свыше 95%.

```python
# Сортируем SKU внутри каждого периода по убыванию выручки.
# При одинаковой выручке — по коду товара для воспроизводимости.
abc = (
    sku_structure.reset_index()
    .sort_values(["Period", "revenue", "StockCode"],
                 ascending=[True, False, True])
    .copy()
)

# Накопленная доля выручки — отдельно в каждом периоде
abc["cumulative_share_pct"] = (
    abc.groupby("Period", observed=True)["revenue_share_pct"].cumsum()
)

# Классы по границам 80% и 95%
abc["ABC"] = np.select(
    [abc["cumulative_share_pct"] <= 80,
     abc["cumulative_share_pct"] <= 95],
    ["A", "B"],
    default="C"
)

# Итог: сколько товаров в каждом классе и какую долю продаж они дают
abc_summary = (
    abc.groupby(["Period", "ABC"], observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        revenue=("revenue", "sum"),
        revenue_share_pct=("revenue_share_pct", "sum")
    )
)

print("ABC ПО ВЫРУЧКЕ ПРОДАЖ ДО ВОЗВРАТОВ")
display(abc_summary.round(2))
```

- **Группа A:** около 80% выручки обеспечивают **919 SKU** в первом периоде и **823 SKU** во втором — примерно **22,6% и 21,7%** продаваемого ассортимента.
- **Группа B:** около 15% выручки приходится на **1 072 и 964 SKU** соответственно.
- **Группа C:** более половины продаваемого ассортимента — **2 082 и 2 010 SKU** — обеспечивает лишь около **5% выручки**.

Во втором периоде 80% выручки формирует меньшее число товаров, а их доля в ассортименте немного снизилась. Это согласуется с **небольшим усилением концентрации продаж**.

**Для проекта:** группа C — отправная точка для дальнейшей проверки, но не готовый список на исключение. Нужно учитывать регулярность продаж, новизну, сезонность и роль товаров в корзине. Классификация выполнена **до возвратов**, поэтому значимость отдельных SKU, включая `23166`, требует дополнительной проверки.

### Кривая Парето

## XYZ-анализ вариативности продаж

Сначала построим полный 12-месячный ряд продаж для каждого товара в каждом периоде. Месяцы без продаж заполняем нулями.

Берём только положительные товарные продажи. В каждом периоде рассматриваем те же SKU, которым уже присвоен ABC-класс.

### Построение полных рядов продаж по каждому товару

```python
periods = {
    "2009/10": ("2009-12", "2010-11"),
    "2010/11": ("2010-12", "2011-11"),
}

parts = []

for period, (start, end) in periods.items():
    sales_period = clean.loc[
        ~clean["IsCancel"] & clean["Period"].eq(period)
    ]

    # Полная сетка: каждый SKU периода × 12 месяцев
    skus = sku_period.loc[
        sku_period["Period"].eq(period), "StockCode"
    ]
    months = pd.period_range(start, end, freq="M")

    grid = pd.MultiIndex.from_product(
        [skus, months], names=["StockCode", "Month"]
    )

    # Количество проданных единиц; отсутствующие месяцы — нули
    monthly = (
        sales_period
        .groupby(
            ["StockCode", sales_period["InvoiceDate"].dt.to_period("M")],
            observed=True
        )["Quantity"]
        .sum()
        .rename_axis(["StockCode", "Month"])
        .reindex(grid, fill_value=0)
        .rename("units")
        .reset_index()
        .assign(Period=period)
    )

    parts.append(monthly)

sku_month = pd.concat(parts, ignore_index=True)[
    ["Period", "StockCode", "Month", "units"]
]

# Контроль: каждый SKU имеет ровно 12 разных месяцев
assert not sku_month.duplicated(["Period", "StockCode", "Month"]).any()
assert (
    sku_month.groupby(["Period", "StockCode"])["Month"]
    .nunique().eq(12).all()
)
assert sku_month["units"].sum() == clean.loc[
    ~clean["IsCancel"], "Quantity"
].sum()

# Компактная проверочная таблица
display(
    sku_month.groupby("Period").agg(
        sku_count=("StockCode", "nunique"),
        months=("Month", "nunique"),
        sku_month_rows=("units", "size"),
        zero_sales_months=("units", lambda x: x.eq(0).sum()),
        units_sold=("units", "sum"),
    )
)
```

#### Вывод по подготовке месячных рядов

Создана таблица `sku_month`: **каждому SKU соответствует полный ряд из 12 месяцев** в своём годовом периоде.

- **2009/10:** 4 073 SKU, 48 876 строк «товар × месяц».
- **2010/11:** 3 797 SKU, 45 564 строки.
- Без положительных продаж — **33,6% и 31,9% товаро-месяцев** соответственно. Эти значения заполнены нулями.
- Суммарное количество проданных единиц совпадает с исходной выборкой продаж.

Месяцы без продаж могут отражать отсутствие товара в наличии, недавнее появление, сезонность или редкие покупки — **не обязательно отсутствие спроса**.

### Расчёт показателей вариативности продаж

Используем готовую sku_month. Для каждого SKU в каждом периоде считаем:

active_months — число месяцев с положительными продажами;
mean_qty — среднее месячное количество за все 12 месяцев;
std_qty — стандартное отклонение;
CV — отношение стандартного отклонения к среднему.
Используем ddof=0: описываем вариативность полного набора из 12 месяцев выбранного периода.

```python
xyz_metrics = (
    sku_month.groupby(["Period", "StockCode"], observed=True)
    .agg(
        active_months=("units", lambda x: x.gt(0).sum()),
        mean_qty=("units", "mean"),
        std_qty=("units", lambda x: x.std(ddof=0)),
    )
)

xyz_metrics["CV"] = (
    xyz_metrics["std_qty"] / xyz_metrics["mean_qty"]
)

# У всех рассматриваемых SKU есть продажи хотя бы в одном месяце
assert xyz_metrics["mean_qty"].gt(0).all()
assert np.isfinite(xyz_metrics["CV"]).all()

# Компактная сводка для проверки
xyz_check = (
    xyz_metrics.groupby(level="Period", observed=True)
    .agg(
        sku_count=("CV", "size"),
        median_active_months=("active_months", "median"),
        sold_all_12_months=("active_months", lambda x: x.eq(12).sum()),
        sold_only_1_month=("active_months", lambda x: x.eq(1).sum()),
        median_CV=("CV", "median"),
        min_CV=("CV", "min"),
        max_CV=("CV", "max"),
    )
)

display(xyz_check.round(2))
```

#### Вывод по вариативности продаж

- В обоих периодах медианное число месяцев с продажами — **9 из 12**.
- Продажи во всех 12 месяцах имели **1 233 SKU (30,3%)** в первом периоде и **1 333 SKU (35,1%)** во втором.
- Только в одном месяце продавались **262 и 225 SKU** соответственно.
- Медианный CV снизился с **1,33 до 1,22**. Вариативность продаж у медианного товара стала ниже, однако состав ассортимента менялся — это не доказывает улучшение стабильности одних и тех же SKU.
- Максимальный CV **3,32** ожидаем для товара с продажами только в одном из 12 месяцев.

**При базовых порогах из ТЗ более половины SKU попадут в Z**, поскольку медианный CV превышает 1 в обоих периодах. Это не означает, что товары «плохие»: высокий CV может быть связан с редкими продажами, сезонностью, короткой историей или крупными заказами.

**Получена таблица `xyz_metrics` с активными месяцами, средним месячным количеством, стандартным отклонением и CV. Следующий шаг — присвоить классы X/Y/Z по порогам 0,5 и 1 и посмотреть распределение.**

### Присвоение XYZ-классов

Используем пороги 0,75 и 1,5:
X: CV ≤ 0,75 — относительно низкая вариативность;
Y: 0,75 < CV ≤ 1,5 — умеренная;
Z: CV > 1,5 — высокая.

Пороги 0,5/1,0 из ТЗ ориентированы на ассортимент с непрерывными продажами. В наших данных треть товаро-месяцев — нулевые, а медиана активных месяцев — 9 из 12. Поэтому высокий CV часто отражает прерывистость, а не нестабильность спроса. При базовых порогах 62–67% SKU попадали в Z, и класс перестал различать товары. Пороги 0,75/1,5 дают сбалансированное распределение ~20/38/42.

```python
# Пороги XYZ — единый источник для всего раздела.
# Меняются здесь, применяются везде.
XYZ_X_CUT = 0.75   # граница X/Y
XYZ_Y_CUT = 1.5  # граница Y/Z
```

```python
xyz_metrics["XYZ"] = np.select(
    [xyz_metrics["CV"] <= XYZ_X_CUT,
     xyz_metrics["CV"] <= XYZ_Y_CUT],
    ["X", "Y"],
    default="Z"
)

# Количество и доля SKU каждого класса по периодам
xyz_summary = (
    xyz_metrics.reset_index()
    .groupby(["Period", "XYZ"], observed=True)
    .agg(sku_count=("StockCode", "nunique"))
    .reindex(
        pd.MultiIndex.from_product(
            [["2009/10", "2010/11"], ["X", "Y", "Z"]],
            names=["Period", "XYZ"]
        ),
        fill_value=0
    )
)

xyz_summary["sku_share_pct"] = (
    100 * xyz_summary["sku_count"]
    / xyz_summary.groupby(level="Period")["sku_count"].transform("sum")
)

display(xyz_summary.round(2))
```

Класс Z составляет 43,8% и 37,5% SKU — это осмысленный размер группы, а не свалка. Прежние пороги 0,5/1,0 давали Z = 62–67% и делали класс неразличимым. Товар с продажами менее чем в 6 месяцах при старом пороге неизбежно попадал в Z, даже если в активные месяцы продавался ровно.

Для XYZ-классификации использованы пороги 0,75 и 1,5. Они отделяют редкие, но предсказуемые продажи от действительно нерегулярного спроса и применяются одинаково к обоим периодам. Обоснование порогов — в разделе проверки устойчивости.

### Распределение CV и проверка базовых границ

Посмотрим, как распределены значения CV и где относительно них находятся пороги 0,75 и 1,5. Классификацию пока не меняем.

```python
fig, axes = plt.subplots(
    1, 2, figsize=(12, 4), sharex=True, sharey=True
)

# Одинаковые интервалы для сравнения периодов
bins = np.linspace(0, 3.4, 35)

for ax, period in zip(axes, ["2009/10", "2010/11"]):
    values = xyz_metrics.xs(period, level="Period")["CV"]

    ax.hist(
        values,
        bins=bins,
        weights=np.full(len(values), 100 / len(values)),
        color="steelblue",
        edgecolor="white"
    )

    ax.axvline(XYZ_X_CUT, color="green", linestyle="--",
           label=f"X/Y: {XYZ_X_CUT}")
    ax.axvline(XYZ_Y_CUT, color="darkorange", linestyle="--",
           label=f"Y/Z: {XYZ_Y_CUT}")

    ax.set_title(period)
    ax.set_xlabel("Коэффициент вариации (CV)")
    ax.legend()

axes[0].set_ylabel("Доля SKU в интервале, %")
plt.tight_layout()
plt.show()

# Квантили для числовой оценки распределения
cv_quantiles = (
    xyz_metrics.groupby(level="Period", observed=True)["CV"]
    .quantile([0.10, 0.25, 0.50, 0.75, 0.90])
    .unstack()
)

cv_quantiles.columns = ["P10", "P25", "Медиана", "P75", "P90"]
display(cv_quantiles.round(2))
```

При них распределение становится сбалансированным (~20/38/42), а Z перестаёт быть свалкой. Высокий CV по-прежнему требует анализа причин колебаний, а не автоматического исключения товаров.

### Добавление XYZ-показателей в витрину sku_period


Пороги 0,75 и 1,5 сохраняем. Теперь перенесём рассчитанные показатели из xyz_metrics в товарную витрину, чтобы ABC и XYZ находились в одной таблице для дальнейшего анализа и BI.

Объединяем по периоду и коду товара, а не только по SKU.

```python
xyz_columns = [
    "active_months", "mean_qty", "std_qty", "CV", "XYZ"
]

rows_before = len(sku_period)

# Удаление этих столбцов позволяет безопасно перезапускать ячейку
sku_period = (
    sku_period.drop(columns=xyz_columns, errors="ignore")
    .merge(
        xyz_metrics.reset_index()[
            ["Period", "StockCode"] + xyz_columns
        ],
        on=["Period", "StockCode"],
        how="left",
        validate="one_to_one"
    )
)

# Проверяем, что строки не потерялись и XYZ найден для каждого SKU
assert len(sku_period) == rows_before
assert not sku_period.duplicated(["Period", "StockCode"]).any()
assert sku_period[xyz_columns].notna().all().all()

display(
    sku_period[
        ["Period", "StockCode", "ProductName",
         "revenue", "units", "ABC",
         "active_months", "mean_qty", "std_qty", "CV", "XYZ"]
    ].head(10).round(2)
)

print("Строк в витрине:", len(sku_period))
print("Столбцы:", sku_period.columns.tolist())
```

Витрина sku_period дополнена XYZ-классом и показателями месячной вариативности продаж. Сохранён уровень детализации «SKU × годовой период», число строк не изменилось. В одной таблице теперь доступны вклад товара в выручку (ABC), вариативность продаж (XYZ) и число активных месяцев. Таблица операций clean не изменена.

## Формирование матрицы ABC/XYZ

Теперь посмотрим, как сочетаются вклад товара в выручку и вариативность его продаж. Для каждого периода построим две таблицы:

количество SKU в каждой группе;
доля группы в общей выручке периода.

доли выручки рассчитаны от всей выручки периода, а не внутри строки A/B/C. Поэтому сумма всех девяти ячеек одной денежной матрицы составляет 100% с учётом округления.

Группы CX и CY помогут увидеть товары с небольшим вкладом в выручку, но сравнительно невысокой вариативностью продаж. CZ пока не считаем списком на исключение — причины высокой вариативности ещё предстоит проверить.

```python
# Совместный класс в товарной витрине
sku_period["ABC_XYZ"] = (
    sku_period["ABC"].astype(str)
    + sku_period["XYZ"].astype(str)
)

# Все 9 сочетаний для каждого периода, включая пустые группы
grid = pd.MultiIndex.from_product(
    [
        sku_period["Period"].drop_duplicates(),
        ["A", "B", "C"],
        ["X", "Y", "Z"]
    ],
    names=["Period", "ABC", "XYZ"]
)

abc_xyz_summary = (
    sku_period
    .groupby(["Period", "ABC", "XYZ"], observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        revenue=("revenue", "sum")
    )
    .reindex(grid, fill_value=0)
)

abc_xyz_summary["revenue_share_pct"] = (
    100 * abc_xyz_summary["revenue"]
    / abc_xyz_summary.groupby(level="Period")["revenue"].transform("sum")
)

# Контроль полноты сводки
assert abc_xyz_summary["sku_count"].sum() == len(sku_period)
assert np.isclose(
    abc_xyz_summary["revenue"].sum(),
    sku_period["revenue"].sum()
)

for period in sku_period["Period"].drop_duplicates():
    period_summary = abc_xyz_summary.xs(period, level="Period")

    print(f"\n{period} — количество SKU")
    display(
        period_summary["sku_count"]
        .unstack("XYZ")
        .reindex(index=["A", "B", "C"], columns=["X", "Y", "Z"])
    )

    print(f"{period} — доля в выручке периода, %")
    display(
        period_summary["revenue_share_pct"]
        .unstack("XYZ")
        .reindex(index=["A", "B", "C"], columns=["X", "Y", "Z"])
        .round(2)
    )

# Справочник имён: одно имя на StockCode
name_map = (
    clean.dropna(subset=["Description"])
    .drop_duplicates("StockCode")
    .set_index("StockCode")["Description"]
)
# ============================================================
# SKU только с возвратами в периоде
# ============================================================
# В ABC/XYZ они не участвуют: положительных продаж в периоде нет.
# Но они нужны для сверки сальдо и сценариев вывода: товар,
# у которого в периоде только возвраты, — прямой кандидат на вывод.
#
# Логика согласована с ETL (features.py) и базовым набором
# дашборда «Класс CZ»: база = CZ ∪ SKU только с возвратами.
returns_only = (
    clean.loc[clean["IsCancel"]]
    .groupby(["Period", "StockCode"], observed=True)
    .size()
    .reset_index(name="_has_returns")
    [["Period", "StockCode"]]
)

existing = sku_period[["Period", "StockCode"]]
missing = (
    returns_only
    .merge(existing, on=["Period", "StockCode"],
           how="left", indicator=True)
    .query("_merge == 'left_only'")
    .drop(columns="_merge")
)

if len(missing) > 0:
    missing = missing.assign(
        ProductName=missing["StockCode"].map(name_map).fillna("Нет описания"),
        revenue=0.0,
        units=0,
        orders=0,
        revenue_share_pct=np.nan,
        cumulative_share_pct=np.nan,
        ABC=pd.NA,
        active_months=pd.NA,
        mean_qty=np.nan,
        std_qty=np.nan,
        CV=np.nan,
        XYZ=pd.NA,
        ABC_XYZ=pd.NA,
    )
    sku_period = (
        pd.concat([sku_period, missing], ignore_index=True)
        .sort_values(["Period", "StockCode"])
        .reset_index(drop=True)
    )
    print(f"Добавлено SKU только с возвратами: {len(missing)}")

print("Строк в витрине:", len(sku_period))
display(
    sku_period.groupby("Period", observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        with_sales=("revenue", lambda x: (x > 0).sum()),
        returns_only=("ABC_XYZ", lambda x: x.isna().sum()),
    )
)
```

## Динамика продаж для адаптированной BCG

Для каждого товара рассчитаем:

выручку и количество в обоих периодах;
абсолютное изменение;
процентное изменение, только если в первом периоде была положительная база.

```python
# Одна строка на SKU, показатели двух периодов — в столбцах
sku_dynamics = (
    sku_period.pivot(
        index="StockCode",
        columns="Period",
        values=["revenue", "units"]
    )
    .reindex(
        columns=pd.MultiIndex.from_product([
            ["revenue", "units"],
            ["2009/10", "2010/11"]
        ])
    )
    .fillna(0)
)

sku_dynamics.columns = [
    "revenue_p1", "revenue_p2",
    "units_p1", "units_p2"
]

for metric in ["revenue", "units"]:
    base = sku_dynamics[f"{metric}_p1"]
    current = sku_dynamics[f"{metric}_p2"]

    sku_dynamics[f"{metric}_change"] = current - base

    # При нулевой базе процентное изменение не определено
    sku_dynamics[f"{metric}_growth_pct"] = (
        100 * (current - base) / base.where(base.gt(0))
    )

# Наличие положительных продаж в каждом периоде
has_p1 = sku_dynamics["units_p1"].gt(0)
has_p2 = sku_dynamics["units_p2"].gt(0)

sku_dynamics["sales_history"] = np.select(
    [
        has_p1 & has_p2,
        ~has_p1 & has_p2,
        has_p1 & ~has_p2
    ],
    [
        "Продажи в обоих периодах",
        "Продажи только во втором",
        "Продажи только в первом"
    ],
    default="Нет положительных продаж"
)

# Контроль сохранения выручки и количества
for period, suffix in [("2009/10", "p1"), ("2010/11", "p2")]:
    for metric in ["revenue", "units"]:
        assert np.isclose(
            sku_dynamics[f"{metric}_{suffix}"].sum(),
            sku_period.loc[
                sku_period["Period"].eq(period), metric
            ].sum()
        )

# Товары только с возвратами не имеют положительных продаж —
# это допустимо, они попадают в отбор через negative_amount.
# Проверяем только: если у SKU нет продаж ни в одном периоде,
# то и выручка у него должна быть нулевой.
no_sales = ~(has_p1 | has_p2)
if no_sales.any():
    zero_rev = sku_period.loc[
        sku_period["StockCode"].isin(sku_dynamics.index[no_sales]),
        "revenue",
    ].max()
    assert zero_rev == 0, (
        "У SKU без положительных продаж ненулевая выручка — "
        "проверь связь sku_dynamics и sku_period"
    )
print(f"SKU только с возвратами: {int(no_sales.sum())}")

dynamics_check = (
    sku_dynamics.groupby("sales_history")
    .agg(
        sku_count=("units_p1", "size"),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        units_p1=("units_p1", "sum"),
        units_p2=("units_p2", "sum")
    )
)

display(dynamics_check.round(2))
```

В обоих периодах продажи зарегистрированы у 3 146 SKU. Общая выручка увеличилась на 2,51%, тогда как выручка сопоставимого набора товаров снизилась на 12,78%, а количество проданных единиц — на 18,06%. Общий денежный рост обеспечен вкладом SKU с продажами только во втором периоде. Поэтому дальнейшая оценка динамики проводится с разделением товаров по истории продаж; отсутствие продаж не трактуется автоматически как появление или вывод товара.

### Проверка темпов роста и малой базы


Для 3 231 SKU с продажами в обоих периодах посмотрим:

Распределение изменений выручки и количества.
Насколько процентный рост зависит от выручки первого периода.
Пока не присваиваем категории адаптированной BCG.

```python
# Только товары с продажами в обоих периодах
common_sku = sku_dynamics.loc[
    sku_dynamics["sales_history"].eq("Продажи в обоих периодах")
].copy()

growth_columns = ["revenue_growth_pct", "units_growth_pct"]

# Проверяем наличие положительной базы и корректность темпов роста
assert common_sku[["revenue_p1", "units_p1"]].gt(0).all().all()
assert np.isfinite(common_sku[growth_columns].to_numpy()).all()

# 1. Распределение индивидуальных темпов роста, %
growth_quantiles = (
    common_sku[growth_columns]
    .quantile([0, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99, 1])
)

growth_quantiles.index = [
    "Минимум", "P10", "P25", "Медиана",
    "P75", "P90", "P95", "P99", "Максимум"
]
growth_quantiles.columns = [
    "Изменение выручки, %",
    "Изменение количества, %"
]

print("РАСПРЕДЕЛЕНИЕ ТЕМПОВ РОСТА")
display(growth_quantiles.round(2))

# 2. Группы по размеру выручки первого периода:
# Q1 — наименьшая база, Q4 — наибольшая.
# При совпадающих границах число групп может быть меньше четырёх.
base_bins = pd.qcut(
    common_sku["revenue_p1"],
    q=4,
    duplicates="drop"
)

common_sku["base_group"] = base_bins.cat.rename_categories(
    [f"Q{i + 1}" for i in range(len(base_bins.cat.categories))]
)

base_check = (
    common_sku.groupby("base_group", observed=True)
    .agg(
        sku_count=("revenue_p1", "size"),
        min_base_revenue=("revenue_p1", "min"),
        max_base_revenue=("revenue_p1", "max"),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        revenue_change=("revenue_change", "sum"),
        median_growth_pct=("revenue_growth_pct", "median"),
        p90_growth_pct=(
            "revenue_growth_pct", lambda x: x.quantile(0.90)
        ),
        growing_sku_pct=(
            "revenue_growth_pct", lambda x: 100 * x.gt(0).mean()
        )
    )
)

# Рост суммарной выручки группы — не среднее процентов по SKU
base_check["group_revenue_growth_pct"] = (
    100 * base_check["revenue_change"] / base_check["revenue_p1"]
)

print("ДИНАМИКА В ЗАВИСИМОСТИ ОТ НАЧАЛЬНОЙ ВЫРУЧКИ")
display(base_check.round(2))
```

Распределение темпов роста асимметрично и содержит экстремальные положительные значения. В группе с наименьшей начальной выручкой процентные приросты существенно выше, что указывает на необходимость отдельной проверки эффекта малой базы. При этом медианное изменение выручки отрицательно во всех квартильных группах. Основной абсолютный спад приходится на товары с высокой начальной выручкой. Поэтому дальнейшая оценка роста будет учитывать не только процентное изменение, но и абсолютный вклад, размер базы и историю продаж.

### Проверка SKU с максимальным процентным ростом

Посмотрим 20 товаров с наибольшим ростом выручки: насколько велика начальная база, сколько денег добавилось и в скольких месяцах были продажи.

```python
# Активные месяцы в каждом периоде
active_by_period = (
    sku_period.pivot(
        index="StockCode",
        columns="Period",
        values="active_months"
    )
    .reindex(columns=["2009/10", "2010/11"])
    .rename(columns={
        "2009/10": "active_months_p1",
        "2010/11": "active_months_p2"
    })
)

# Название для отображения: из второго периода,
# при отсутствии — из первого
names_by_period = (
    sku_period.pivot(
        index="StockCode",
        columns="Period",
        values="ProductName"
    )
    .reindex(columns=["2009/10", "2010/11"])
)

product_names = (
    names_by_period["2010/11"]
    .combine_first(names_by_period["2009/10"])
    .rename("ProductName")
)

# Только SKU с продажами в обоих периодах
growth_audit = (
    sku_dynamics.loc[
        sku_dynamics["sales_history"].eq("Продажи в обоих периодах")
    ]
    .join(active_by_period, validate="one_to_one")
    .join(product_names, validate="one_to_one")
)

assert growth_audit[
    ["active_months_p1", "active_months_p2"]
].notna().all().all()

audit_columns = [
    "ProductName",
    "revenue_p1", "revenue_p2",
    "revenue_change", "revenue_growth_pct",
    "units_p1", "units_p2",
    "units_growth_pct",
    "active_months_p1", "active_months_p2"
]

top_growth_audit = (
    growth_audit
    .sort_values(
        ["revenue_growth_pct", "revenue_change"],
        ascending=[False, False]
    )
    .head(20)
    [audit_columns]
)

with pd.option_context(
    "display.max_rows", 30,
    "display.max_columns", None,
    "display.max_colwidth", 45
):
    display(top_growth_audit.round(2))
```

У всех 20 SKU с максимальным процентным ростом выручки продажи в базовом периоде наблюдались лишь в одном или двух месяцах. Во втором периоде число активных месяцев увеличилось до 5–12. Экстремальные темпы роста поэтому требуют оговорки о малой базе и различиях в активности продаж. При этом абсолютный денежный прирост существенно различается между товарами. Для дальнейшей классификации сохраняются исходные темпы роста, но их интерпретация дополняется абсолютным изменением выручки и числом активных месяцев.



### Динамика по числу активных месяцев базового периода

Проверим все сопоставимые SKU, а не только лидеров роста. Пока используем точное число активных месяцев — от 1 до 12, без произвольного разделения на «короткую» и «достаточную» историю.

```python
# growth_audit — общие SKU с динамикой и активными месяцами
activity_check = (
    growth_audit
    .groupby("active_months_p1", observed=True)
    .agg(
        sku_count=("revenue_p1", "size"),
        median_base_revenue=("revenue_p1", "median"),
        median_active_months_p2=("active_months_p2", "median"),
        more_active_months_pct=(
            "active_months_p2",
            lambda x: 100 * (
                x > growth_audit.loc[x.index, "active_months_p1"]
            ).mean()
        ),
        median_growth_pct=("revenue_growth_pct", "median"),
        p90_growth_pct=(
            "revenue_growth_pct", lambda x: x.quantile(0.90)
        ),
        growing_sku_pct=(
            "revenue_growth_pct", lambda x: 100 * x.gt(0).mean()
        ),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        revenue_change=("revenue_change", "sum")
    )
    .sort_index()
)

# Изменение общей выручки группы
activity_check["group_revenue_growth_pct"] = (
    100 * activity_check["revenue_change"]
    / activity_check["revenue_p1"]
)

activity_check.index = activity_check.index.astype(int)
activity_check.index.name = "Активных месяцев в первом периоде"

assert activity_check["sku_count"].sum() == len(growth_audit)
assert np.isclose(
    activity_check["revenue_change"].sum(),
    growth_audit["revenue_change"].sum()
)

with pd.option_context(
    "display.max_rows", 20,
    "display.max_columns", None
):
    display(activity_check.round(2))
```

**Теперь есть основание выделить SKU с 1–2 активными месяцами в базовом периоде отдельным диагностическим признаком.** Но это не основание исключать их из анализа или считать их рост недостоверным.

|---|---:|---:|---:|

| 2 | +172,24% | 80,85% | 83,69% |
| 3 | +11,84% | 56,21% | 72,55% |
| 4 | −5,55% | 45,24% | 53,17% |

После двух месяцев медианный рост резко уменьшается. Это подтверждает, что выявленная особенность относится **не только к топ-20**, но и к более широкой группе товаров.

Однако здесь одновременно различаются денежная база, число активных месяцев и состав товаров. Поэтому нельзя утверждать, что именно короткая активность *вызвала* высокий рост.

Группа с тремя месяцами — промежуточная

У неё ещё положительная медиана и у большинства товаров увеличилось число активных месяцев, но проценты уже значительно ниже.

 Среди товаров с регулярными продажами в базовом периоде преобладает снижение

- медианное изменение выручки находится между **−28,62% и −36,83%**;
- суммарная выручка каждой группы снизилась;
- рост наблюдается лишь у **18,39–30,14% SKU**.

Особенно значима группа с 12 активными месяцами: **1 202 SKU**, снижение выручки на **1,47 млн, или 27,90%**.

Это товары с продажами в каждом месяце **первого периода**. Таблица не доказывает, что каждый из них продавался все 12 месяцев второго периода или постоянно был в наличии.


### Добавляем признак, но сохраняем показатели

Назовём его буквально: «1–2 активных месяца в базовом периоде», а не «новинка» или «недостоверный рост».

```python
# Признак рассчитываем для SKU с продажами в обоих периодах
growth_audit["limited_base_activity"] = (
    growth_audit["active_months_p1"].le(2)
)

display(
    growth_audit.groupby("limited_base_activity")
    .agg(
        sku_count=("revenue_p1", "size"),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        revenue_change=("revenue_change", "sum")
    )
    .round(2)
)
```

Наиболее высокие темпы роста характерны для SKU с одним или двумя активными месяцами в базовом периоде. В этих группах медианный рост выручки составляет 445,06% и 174,64% и часто сопровождается увеличением числа месяцев с продажами. Для таких товаров введён диагностический признак ограниченной базовой активности. Исходные темпы роста сохраняются, но интерпретируются с учётом абсолютного прироста и истории продаж. Порог в два месяца является рабочим; его чувствительность проверяется при расширении до трёх месяцев.

### Правила адаптированной BCG

Используем **вклад в выручку второго периода** и **изменение выручки между периодами**.

Это внутренняя ассортиментная матрица, **не классическая BCG**: данных о росте рынка и долях конкурентов у нас нет.

#### Рабочие границы

| Ось | Правило | Обоснование |
|---|---|---|
| Значимость | **A — высокая**, B/C — меньшая | Используем уже рассчитанный ABC, не вводя ещё один порог денежного вклада |
| Динамика | **Рост > 0%**, снижение < 0%, без изменения = 0% | Ноль отделяет фактическое увеличение выручки от уменьшения |

**Положительный рост не означает существенный или устойчивый рост.** Пока это простое базовое правило; чувствительность к небольшим изменениям около нуля проверим отдельно.

SKU с 1–2 активными месяцами в первом периоде остаются в матрице, но сохраняют предупреждающий признак. Товары с продажами только в одном периоде пока **не распределяем по квадрантам**.


```python
# ABC второго периода отражает текущий вклад товара в выручку
abc_p2 = (
    sku_period.loc[
        sku_period["Period"].eq("2010/11"),
        ["StockCode", "ABC"]
    ]
    .set_index("StockCode")["ABC"]
    .rename("ABC_p2")
)

bcg_common = growth_audit.join(
    abc_p2,
    validate="one_to_one"
)

assert bcg_common["ABC_p2"].isin(["A", "B", "C"]).all()

# Диагностический признак, а не основание для исключения
bcg_common["limited_base_activity"] = (
    bcg_common["active_months_p1"].le(2)
)

high_contribution = bcg_common["ABC_p2"].eq("A")
change = bcg_common["revenue_change"]

bcg_common["BCG_adapted"] = np.select(
    [
        high_contribution & change.gt(0),
        high_contribution & change.lt(0),
        ~high_contribution & change.gt(0),
        ~high_contribution & change.lt(0),
        high_contribution & change.eq(0),
        ~high_contribution & change.eq(0)
    ],
    [
        "A — рост",
        "A — снижение",
        "B/C — рост",
        "B/C — снижение",
        "A — без изменения",
        "B/C — без изменения"
    ],
    default="Проверить"
)
# Товары без базы в первом периоде — отдельный статус
only_p2_mask = growth_audit["sales_history"].eq("Продажи только во втором")
bcg_common.loc[only_p2_mask, "BCG_adapted"] = "Продажи только во втором"
assert not bcg_common["BCG_adapted"].eq("Проверить").any()

bcg_summary = (
    bcg_common.groupby("BCG_adapted", observed=True)
    .agg(
        sku_count=("revenue_p1", "size"),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        revenue_change=("revenue_change", "sum"),
        median_growth_pct=("revenue_growth_pct", "median"),
        limited_base_sku=("limited_base_activity", "sum")
    )
)

bcg_summary["limited_base_share_pct"] = (
    100 * bcg_summary["limited_base_sku"]
    / bcg_summary["sku_count"]
)

# Доля от ВСЕЙ выручки второго периода,
# включая товары с продажами только во втором периоде
total_revenue_p2 = sku_period.loc[
    sku_period["Period"].eq("2010/11"), "revenue"
].sum()

bcg_summary["revenue_p2_share_pct"] = (
    100 * bcg_summary["revenue_p2"] / total_revenue_p2
)

assert bcg_summary["sku_count"].sum() == len(growth_audit)
assert np.isclose(
    bcg_summary["revenue_p2"].sum(),
    growth_audit["revenue_p2"].sum()
)

with pd.option_context("display.max_columns", None):
    display(bcg_summary.round(2))
```

Адаптированная матрица выделила две приоритетные группы для диагностики снижения: товары текущего класса A, потерявшие 1,09 млн выручки, и товары текущих классов B/C, потерявшие 1,46 млн. Товары с продажами только во втором периоде в матрицу не включены — для них нет базы сравнения, они сохраняют отдельный статус «Продажи только во втором периоде». Это важно для согласования с ETL: там такие SKU не должны получать статус «рост». В растущих группах 18–21% SKU имели ограниченную базовую активность, что требует осторожной интерпретации темпов роста. Поскольку значимость определена по ABC второго периода, далее необходимо проверить переходы между классами: группа B/C со снижением может включать бывшие товары A. Матрица используется для приоритизации анализа, а не для автоматического исключения или расширения ассортимента.

### Переходы ABC между периодами

Проверим, какие товары сохранили класс, какие повысили его, а какие перешли из A в B/C. Анализируем только SKU с продажами в обоих периодах.

Построим:

Матрицу количества SKU по переходам.
Сводку потерь группы «B/C — снижение» с учётом прежнего класса.

```python
# ABC первого периода
abc_p1 = (
    sku_period.loc[
        sku_period["Period"].eq("2009/10"),
        ["StockCode", "ABC"]
    ]
    .set_index("StockCode")["ABC"]
    .rename("ABC_p1")
)

# ABC_p2 уже есть в bcg_common
abc_transitions = (
    bcg_common
    .drop(columns=["ABC_p1"], errors="ignore")
    .join(abc_p1, validate="one_to_one")
)

assert abc_transitions[["ABC_p1", "ABC_p2"]].isin(
    ["A", "B", "C"]
).all().all()

abc_transitions["ABC_transition"] = (
    abc_transitions["ABC_p1"].astype(str)
    + " → "
    + abc_transitions["ABC_p2"].astype(str)
)

# 1. Матрица переходов: строки — первый период,
# столбцы — второй
transition_counts = (
    pd.crosstab(
        abc_transitions["ABC_p1"],
        abc_transitions["ABC_p2"]
    )
    .reindex(
        index=["A", "B", "C"],
        columns=["A", "B", "C"],
        fill_value=0
    )
)

# Доли переходов внутри каждого исходного класса
transition_shares = (
    transition_counts
    .div(transition_counts.sum(axis=1).replace(0, np.nan), axis=0)
    .mul(100)
)

assert transition_counts.to_numpy().sum() == len(bcg_common)

print("ПЕРЕХОДЫ ABC — КОЛИЧЕСТВО SKU")
display(transition_counts)

print("ПЕРЕХОДЫ ABC — % ОТ ИСХОДНОГО КЛАССА")
display(transition_shares.round(2))

# 2. Из чего складывается снижение текущих B/C
bc_decline = abc_transitions.loc[
    abc_transitions["BCG_adapted"].eq("B/C — снижение")
]

decline_by_transition = (
    bc_decline.groupby("ABC_transition", observed=True)
    .agg(
        sku_count=("revenue_p1", "size"),
        revenue_p1=("revenue_p1", "sum"),
        revenue_p2=("revenue_p2", "sum"),
        revenue_change=("revenue_change", "sum")
    )
)

# Положительная величина потери выручки
decline_by_transition["revenue_loss"] = (
    -decline_by_transition["revenue_change"]
)

decline_by_transition["loss_share_pct"] = (
    100 * decline_by_transition["revenue_loss"]
    / decline_by_transition["revenue_loss"].sum()
)

decline_by_transition = decline_by_transition.sort_values(
    "revenue_loss", ascending=False
)

assert decline_by_transition["sku_count"].sum() == len(bc_decline)
assert np.isclose(
    decline_by_transition["revenue_change"].sum(),
    bc_decline["revenue_change"].sum()
)

print("ГРУППА «B/C — СНИЖЕНИЕ»: ПОТЕРИ ПО ПЕРЕХОДАМ")
with pd.option_context("display.max_columns", None):
    display(decline_by_transition.round(2))
```

Анализ переходов ABC показал, что 37,21% сопоставимых товаров исходного класса A перешли в B/C. Бывшие товары A со снижением выручки сформировали 60,21% потерь группы «B/C — снижение», а вместе с переходом B → C — 83,63%. Поэтому текущего ABC-класса недостаточно для определения приоритетов: необходимо учитывать прежнюю значимость и абсолютную потерю выручки. Большинство исходных товаров C сохранили класс, однако это не является основанием для исключения без дополнительной оценки.

### Товары с наибольшими потерями: переходы ABC + текущий XYZ

Теперь соберём список SKU для диагностики снижения продаж. Включим все сопоставимые товары со снижением выручки — в том числе сохранившие класс A, чтобы не пропустить крупные потери.

Ранжируем по абсолютной потере выручки, а XYZ используем как дополнительную характеристику, не как причину снижения.

```python
# XYZ и показатели вариативности второго периода
xyz_p2 = (
    sku_period.loc[
        sku_period["Period"].eq("2010/11"),
        ["StockCode", "XYZ", "CV"]
    ]
    .set_index("StockCode")
    .rename(columns={
        "XYZ": "XYZ_p2",
        "CV": "CV_p2"
    })
)

decline_audit = (
    abc_transitions.loc[
        abc_transitions["revenue_change"].lt(0)
    ]
    .copy()
    .join(xyz_p2, validate="one_to_one")
)

assert decline_audit["XYZ_p2"].isin(["X", "Y", "Z"]).all()

decline_audit["revenue_loss"] = -decline_audit["revenue_change"]

# 1. Денежные потери по переходам ABC и текущему XYZ
loss_by_xyz = (
    decline_audit
    .groupby(["ABC_transition", "XYZ_p2"], observed=True)["revenue_loss"]
    .sum()
    .unstack("XYZ_p2", fill_value=0)
    .reindex(columns=["X", "Y", "Z"], fill_value=0)
)

loss_by_xyz["total_loss"] = loss_by_xyz[["X", "Y", "Z"]].sum(axis=1)
loss_by_xyz = loss_by_xyz.sort_values("total_loss", ascending=False)

assert np.isclose(
    loss_by_xyz["total_loss"].sum(),
    decline_audit["revenue_loss"].sum()
)

print("ПОТЕРИ ВЫРУЧКИ: ПЕРЕХОД ABC × XYZ ВТОРОГО ПЕРИОДА")
display(loss_by_xyz.round(2))

# 2. Рейтинг всех сопоставимых SKU со снижением
decline_audit = decline_audit.sort_values(
    "revenue_loss", ascending=False
)

total_decline_loss = decline_audit["revenue_loss"].sum()

decline_audit["loss_share_pct"] = (
    100 * decline_audit["revenue_loss"] / total_decline_loss
)

decline_audit["cumulative_loss_share_pct"] = (
    decline_audit["loss_share_pct"].cumsum()
)

top_decline = decline_audit.head(20)

print("ТОП-20 SKU ПО АБСОЛЮТНОЙ ПОТЕРЕ ВЫРУЧКИ")
with pd.option_context(
    "display.max_columns", None,
    "display.max_rows", 25,
    "display.max_colwidth", 45
):
    display(
        top_decline[[
            "ProductName",
            "ABC_transition", "XYZ_p2", "CV_p2",
            "revenue_p1", "revenue_p2",
            "revenue_loss", "revenue_growth_pct",
            "units_growth_pct",
            "active_months_p1", "active_months_p2",
            "cumulative_loss_share_pct"
        ]].round(2)
    )

print(
    "Доля топ-20 в сумме потерь всех сопоставимых SKU со снижением: "
    f"{top_decline['revenue_loss'].sum() / total_decline_loss:.2%}"
)
```


- **У сохранивших класс A около 84% потерь приходится на X/Y.** Значит, снижение выручки возможно и при сравнительно регулярных продажах.
- **При переходах A → C и B → C преобладают потери в Z** — около 87% и 75% соответственно.
- **Активность не гарантирует объём:** SKU 21622 продавался все 12 месяцев обоих периодов, но потерял 89% выручки.
- **Топ-20 объясняет только 13,22% потерь** — проблема не ограничивается несколькими товарами.

### Формирование месячной витрины sku_dynamics

Назначение: подготовить для Power BI помесячные продажи товаров с возможностью фильтрации по ABC/XYZ, группе BCG и истории продаж.

Детализация — одна строка на SKU × месяц. Используем готовую сетку sku_month, добавляем выручку из clean и результаты классификаций.

BCG заполняем только для второго периода. Годовые денежные показатели и проценты роста сюда не переносим, чтобы избежать их повторного суммирования.

```python
# Сохраняем прежнюю годовую таблицу сравнения под отдельным именем.
# Условие позволяет повторно запускать эту ячейку.
if "Month" not in sku_dynamics.columns:
    sku_comparison = sku_dynamics.copy()

# 1. Месячная выручка положительных товарных продаж
monthly_revenue = (
    clean.loc[~clean["IsCancel"]]
    .assign(Month=lambda x: x["InvoiceDate"].dt.to_period("M"))
    .groupby(
        ["Period", "StockCode", "Month"], observed=True
    )["Amount"]
    .sum()
    .rename("revenue")
    .reset_index()
)

# 2. Классы соответствующего годового периода
annual_classes = sku_period[
    ["Period", "StockCode", "ABC", "XYZ", "ABC_XYZ"]
]

# 3. BCG сопоставимых товаров — только для второго периода
bcg_labels = (
    bcg_common[["BCG_adapted", "limited_base_activity"]]
    .reset_index()
    .assign(Period="2010/11")
)

# 4. Собираем месячную витрину
sku_dynamics = (
    sku_month.copy()
    .merge(
        monthly_revenue,
        on=["Period", "StockCode", "Month"],
        how="left",
        validate="one_to_one"
    )
    .merge(
        annual_classes,
        on=["Period", "StockCode"],
        how="left",
        validate="many_to_one"
    )
    .merge(
        bcg_labels,
        on=["Period", "StockCode"],
        how="left",
        validate="many_to_one"
    )
)

sku_dynamics["revenue"] = sku_dynamics["revenue"].fillna(0)

sku_dynamics["sales_history"] = (
    sku_dynamics["StockCode"]
    .map(sku_comparison["sales_history"])
)

# Товары без продаж в первом периоде — отдельный статус,
# а не подтверждённые новинки и не квадрант BCG
only_second = (
    sku_dynamics["Period"].eq("2010/11")
    & sku_dynamics["sales_history"].eq("Продажи только во втором")
)

sku_dynamics.loc[only_second, "BCG_adapted"] = (
    "Продажи только во втором"
)

sku_dynamics["limited_base_activity"] = (
    sku_dynamics["limited_base_activity"].astype("boolean")
)

# Первый день месяца — удобный тип даты для Power BI
sku_dynamics["Month"] = (
    sku_dynamics["Month"].dt.to_timestamp()
)

sku_dynamics = sku_dynamics.sort_values(
    ["Period", "StockCode", "Month"]
).reset_index(drop=True)

# 5. Контроль структуры и годовых сумм
assert len(sku_dynamics) == len(sku_month)
assert not sku_dynamics.duplicated(
    ["Period", "StockCode", "Month"]
).any()

assert sku_dynamics[
    ["ABC", "XYZ", "ABC_XYZ", "sales_history"]
].notna().all().all()

actual = (
    sku_dynamics.groupby(
        ["Period", "StockCode"], observed=True
    )[["revenue", "units"]].sum()
)

expected = (
    sku_period.set_index(["Period", "StockCode"])
    [["revenue", "units"]]
    .reindex(actual.index)
)

assert np.allclose(actual.to_numpy(), expected.to_numpy())

is_p2 = sku_dynamics["Period"].eq("2010/11")
assert sku_dynamics.loc[is_p2, "BCG_adapted"].notna().all()
assert sku_dynamics.loc[~is_p2, "BCG_adapted"].isna().all()

display(
    sku_dynamics.groupby("Period", observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        rows=("Month", "size"),
        months=("Month", "nunique"),
        revenue=("revenue", "sum"),
        units=("units", "sum")
    )
)
```

Витрина создана: 97 512 строк на уровне «SKU × месяц».

2009/10: 4 221 SKU, 50 652 строки.
2010/11: 3 905 SKU, 46 860 строк.
Каждому представленному SKU соответствует 12 месяцев, включая месяцы без продаж.
Выручка и количество совпадают с годовыми показателями sku_period; проверки прошли.
В Power BI таблица будет использоваться для месячных графиков, сравнения периодов и фильтрации по ABC/XYZ и BCG. Годовые показатели и сценарии отбора остаются в sku_period.

### Главный вывод: общий рост выручки скрывает снижение продаж сопоставимого ассортимента.

Выручка магазина выросла на **2,51%**, но у **3 146 SKU с продажами в обоих периодах** она снизилась на **12,78%**. Рост обеспечили **651 SKU с продажами только во втором периоде**, принёсшие **2,05 млн GBP**.

При этом **60,21% потерь группы «B/C — снижение» приходится на бывшие товары A**. Поэтому низкий текущий класс не всегда означает, что товар исторически малозначим.

BCG дополняет ABC/XYZ динамикой — помогает отличать растущие товары от снижающихся и видеть потерю прежней значимости. Но снижение продаж само по себе не является основанием для сокращения SKU.

## RFM-анализ покупателей

Задача: определить ценных и регулярно покупающих клиентов, чтобы затем учитывать востребованные ими товары при отборе кандидатов на сокращение.

Начнём с расчёта трёх показателей, без присвоения сегментов:

Recency — дней с последней покупки;
Frequency — число уникальных заказов;
Monetary — выручка положительных покупок до возвратов.
Используем только продажи с известным Customer ID. Каждый год оцениваем отдельно.

```python
# Продажи с известным покупателем
customer_sales = clean.loc[
    ~clean["IsCancel"] & clean["Customer ID"].notna()
].copy()

# Даты оценки — сразу после окончания годового окна
assessment_dates = {
    "2009/10": pd.Timestamp("2010-12-01"),
    "2010/11": pd.Timestamp("2011-12-01")
}

parts = []

for period, assessment_date in assessment_dates.items():
    data = customer_sales.loc[
        customer_sales["Period"].eq(period)
    ]

    table = (
        data.groupby("Customer ID", observed=True)
        .agg(
            last_purchase=("InvoiceDate", "max"),
            Frequency=("Invoice", "nunique"),
            Monetary=("Amount", "sum")
        )
        .reset_index()
    )

    table["Recency"] = (
        assessment_date - table["last_purchase"].dt.normalize()
    ).dt.days

    table["Period"] = period
    table["assessment_date"] = assessment_date
    parts.append(table)

# Витрина «покупатель × годовой период»
rfm = pd.concat(parts, ignore_index=True)[
    ["Period", "Customer ID", "assessment_date",
     "last_purchase", "Recency", "Frequency", "Monetary"]
]

assert not rfm.duplicated(["Period", "Customer ID"]).any()
assert rfm[["Recency", "Frequency", "Monetary"]].gt(0).all().all()

# Сверка денежных сумм отдельно по периодам
expected = customer_sales.groupby(
    "Period", observed=True
)["Amount"].sum()

actual = rfm.groupby("Period")["Monetary"].sum()

assert np.allclose(
    actual.reindex(expected.index).to_numpy(),
    expected.to_numpy()
)

# Покрытие клиентскими данными — по выручке
all_revenue = (
    clean.loc[~clean["IsCancel"]]
    .groupby("Period", observed=True)["Amount"].sum()
)

rfm_summary = (
    rfm.groupby("Period")
    .agg(
        customers=("Customer ID", "nunique"),
        median_recency=("Recency", "median"),
        median_frequency=("Frequency", "median"),
        median_monetary=("Monetary", "median"),
        known_customer_revenue=("Monetary", "sum")
    )
)

rfm_summary["revenue_coverage_pct"] = (
    100 * rfm_summary["known_customer_revenue"] / all_revenue
)

display(rfm_summary.round(2))

print("РАСПРЕДЕЛЕНИЕ RFM")
display(
    rfm.groupby("Period")[["Recency", "Frequency", "Monetary"]]
    .quantile([0.25, 0.50, 0.75, 0.90])
    .round(2)
)
```


- Медианный покупатель совершает **2 заказа в год**.
- Медианная сумма покупок снизилась с **679,80 до 646,42 GBP**.
- Клиентский анализ охватывает **85,39% выручки второго периода**. Остальные продажи сохраняем в товарном анализе, но связать их с клиентскими сегментами нельзя.

Различаем:
- **Показатели:** Recency — дни, Frequency — заказы, Monetary — сумма покупок.
- **Баллы:** `R_score`, `F_score`, `M_score` — оценки от 1 до 3.

| Балл | R — давность | F — частота | M — сумма |
|---|---|---|---|
| **3** | Покупал недавно | 5+ заказов | Высокая сумма |
| **2** | Средняя давность | 2–4 заказа | Средняя сумма |
| **1** | Покупал давно | 1 заказ | Низкая сумма |

Таким образом, **333** — покупатель с недавней покупкой, высокой частотой и суммой. Это код сочетания признаков, а не число для арифметического сравнения.

 Как выберем границы

**Recency и Monetary** — по 33-му и 67-му процентилям внутри каждого периода. Одинаковые значения получают одинаковый балл (`rank(method="dense")`), чтобы клиенты с равной давностью или суммой не расходились по разным группам.

**Frequency** — по явным границам, а не по квантилям:
- **1 заказ** → F=1 (разовый покупатель);
- **2–4 заказа** → F=2 (повторный);
- **5+ заказов** → F=3 (регулярный).

Это гарантирует, что **клиент с одним заказом никогда не получит F=2 или F=3** — независимо от размера группы. Для сегментов, которые защищают товары от вывода (Loyal, Champions), это критично: одноразовая покупка не должна попадать в «повторных покупателей».

 Дополнительно для Loyal и Champions

Кроме баллов, в сегментацию добавлено **явное условие повторной покупки**: `Frequency ≥ 2`. Даже если клиент попал в F=2 по квантилям, без второго заказа он не станет Loyal или Champions.

#### Расчёт баллов

```python
parts = []

for period, group in rfm.groupby("Period", observed=True):
    group = group.copy()

    # R: квантили, но одинаковые значения → одинаковый ранг
    group["R_score"] = pd.qcut(
        group["Recency"].rank(method="dense"),
        q=3, labels=[3, 2, 1], duplicates="drop"
    ).astype(int)

    # F: явные границы по числу заказов
    group["F_score"] = np.select(
        [group["Frequency"] >= 5,
         group["Frequency"] >= 2],
        [3, 2],
        default=1
    )

    # M: квантили, dense
    group["M_score"] = pd.qcut(
        group["Monetary"].rank(method="dense"),
        q=3, labels=[1, 2, 3], duplicates="drop"
    ).astype(int)

    parts.append(group)

rfm = pd.concat(parts, ignore_index=True)

r = rfm["R_score"]
f = rfm["F_score"]
m = rfm["M_score"]
freq = rfm["Frequency"]

rfm["RFM_segment"] = np.select(
    [
        r.eq(3) & f.eq(3) & m.eq(3),
        r.ge(2) & f.ge(2) & freq.ge(2),          # ← Loyal только с 2+ заказами
        r.eq(3) & f.eq(1),
        r.eq(1) & f.ge(2),
        r.eq(1)
    ],
    [
        "Champions — лучшие по RFM",
        "Loyal — повторные покупатели",
        "Recent single — недавняя разовая покупка",
        "At Risk — давние повторные покупатели",
        "Inactive — давняя разовая покупка"
    ],
    default="Regular — остальные"
)
```

```python
# Смотрим, есть ли в Loyal клиенты с одним заказом
loyal = rfm[rfm["RFM_segment"] == "Loyal — повторные покупатели"]
print("Клиентов в Loyal:", len(loyal))
print("Минимум заказов в Loyal:", loyal["Frequency"].min())
print("Одноразовых в Loyal:", (loyal["Frequency"] == 1).sum())
```

```python
# Сводка по сегментам: сколько клиентов, их средние показатели и вклад
segment_summary = (
    rfm.groupby(["Period", "RFM_segment"], observed=True)
    .agg(
        Клиентов=("Customer ID", "nunique"),
        Выручка=("Monetary", "sum"),
        Ср_Recency=("Recency", "mean"),
        Ср_заказов=("Frequency", "mean"),
        Медиана_заказов=("Frequency", "median"),
        Мин_заказов=("Frequency", "min"),
        Макс_заказов=("Frequency", "max"),
        Ср_чек=("Monetary", "mean"),
    )
)

# Доли клиентов и выручки внутри периода
segment_summary["Доля_клиентов_%"] = (
    100 * segment_summary["Клиентов"]
    / segment_summary.groupby(level="Period")["Клиентов"].transform("sum")
).round(1)

segment_summary["Доля_выручки_%"] = (
    100 * segment_summary["Выручка"]
    / segment_summary.groupby(level="Period")["Выручка"].transform("sum")
).round(1)

# Округление для читаемости
segment_summary["Выручка"] = segment_summary["Выручка"].round(0)
segment_summary["Ср_Recency"] = segment_summary["Ср_Recency"].round(0)
segment_summary["Ср_заказов"] = segment_summary["Ср_заказов"].round(1)
segment_summary["Ср_чек"] = segment_summary["Ср_чек"].round(0)

# Порядок сегментов — по смыслу
order = [
    "Champions — лучшие по RFM",
    "Loyal — повторные покупатели",
    "Recent single — недавняя разовая покупка",
    "At Risk — давние повторные покупатели",
    "Regular — остальные",
    "Inactive — давняя разовая покупка",
]

segment_summary = (
    segment_summary
    .reindex(
        pd.MultiIndex.from_product(
            [sorted(rfm["Period"].unique()), order],
            names=["Period", "RFM_segment"]
        )
    )
    .fillna(0)
)

display(segment_summary)
```

Во втором периоде Champions — 14,50% известных покупателей, они дают 57,00% их выручки. Вместе с Loyal две группы охватывают 50,80% покупателей и формируют 85,00% выручки продаж с известным клиентом.

Доля выручки Champions выросла с 52,90% до 57,00%, доля Loyal снизилась с 31,60% до 28,00%. Доля Inactive (давняя разовая покупка) выросла с 18,40% до 20,00%.

### Связь RFM-сегментов с ассортиментом
Цель: определить, какую часть продаж каждого товара обеспечивают Champions и Loyal. Добавим показатели в sku_period, сохранив детализацию «SKU × годовой период».

Клиенту присваиваем сегмент того же периода, в котором совершена покупка. Это ретроспективная оценка, а не прогноз поведения.

```python
# 1. Присоединяем сегменты к положительным продажам
#    с известным покупателем
customer_sku_sales = (
    clean.loc[
        ~clean["IsCancel"] & clean["Customer ID"].notna(),
        ["Period", "StockCode", "Customer ID", "Amount"]
    ]
    .merge(
        rfm[["Period", "Customer ID", "RFM_segment"]],
        on=["Period", "Customer ID"],
        how="left",
        validate="many_to_one"
    )
)

assert customer_sku_sales["RFM_segment"].notna().all()

# Champions и Loyal — значимые сегменты для проверки рисков
key_segments = [
    "Champions — лучшие по RFM",
    "Loyal — повторные покупатели"
]

is_key = customer_sku_sales["RFM_segment"].isin(key_segments)

customer_sku_sales["key_customer_id"] = (
    customer_sku_sales["Customer ID"].where(is_key)
)
customer_sku_sales["key_revenue"] = (
    customer_sku_sales["Amount"].where(is_key, 0)
)

# 2. Агрегируем на уровне SKU × период
sku_client_metrics = (
    customer_sku_sales
    .groupby(["Period", "StockCode"], observed=True)
    .agg(
        known_customers=("Customer ID", "nunique"),
        known_customer_revenue=("Amount", "sum"),
        key_customers=("key_customer_id", "nunique"),
        key_customer_revenue=("key_revenue", "sum")
    )
    .reset_index()
)

# 3. Добавляем показатели в основную годовую витрину
#    Повторный запуск не создаёт дублирующихся столбцов
count_columns = ["known_customers", "key_customers"]
amount_columns = ["known_customer_revenue", "key_customer_revenue"]
share_columns = [
    "customer_revenue_coverage_pct",
    "key_customer_revenue_share_pct",
    "key_customer_share_pct"
]

rows_before = len(sku_period)

sku_period = (
    sku_period
    .drop(
        columns=count_columns + amount_columns + share_columns,
        errors="ignore"
    )
    .merge(
        sku_client_metrics,
        on=["Period", "StockCode"],
        how="left",
        validate="one_to_one"
    )
)

sku_period[count_columns] = (
    sku_period[count_columns].fillna(0).astype("int64")
)
sku_period[amount_columns] = (
    sku_period[amount_columns].fillna(0)
)

# Знаменатель — вся выручка SKU, включая неизвестных клиентов
sku_period["customer_revenue_coverage_pct"] = (
    100 * sku_period["known_customer_revenue"]
    / sku_period["revenue"]
)

sku_period["key_customer_revenue_share_pct"] = (
    100 * sku_period["key_customer_revenue"]
    / sku_period["revenue"]
)

# Знаменатель — только известные покупатели SKU.
# Если их нет, долю оставляем неопределённой, а не равной нулю.
sku_period["key_customer_share_pct"] = (
    100 * sku_period["key_customers"]
    / sku_period["known_customers"].replace(0, np.nan)
)

# 4. Проверки
assert len(sku_period) == rows_before
assert not sku_period.duplicated(["Period", "StockCode"]).any()
assert sku_period["key_customers"].le(
    sku_period["known_customers"]
).all()

assert (
    sku_period["key_customer_revenue"]
    <= sku_period["known_customer_revenue"] + 1e-6
).all()
assert (
    sku_period["known_customer_revenue"]
    <= sku_period["revenue"] + 1e-6
).all()

actual = (
    sku_period.groupby("Period", observed=True)
    ["known_customer_revenue"].sum()
)
expected = rfm.groupby("Period", observed=True)["Monetary"].sum()

assert np.allclose(
    actual.reindex(expected.index).to_numpy(),
    expected.to_numpy()
)

# 5. Краткая сводка: охват товаров значимыми сегментами
client_coverage_summary = (
    sku_period.assign(
        has_known_customers=sku_period["known_customers"].gt(0),
        has_key_customers=sku_period["key_customers"].gt(0)
    )
    .groupby("Period", observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        sku_with_known_customers=("has_known_customers", "sum"),
        sku_with_key_customers=("has_key_customers", "sum")
    )
)

client_coverage_summary["sku_with_key_customers_pct"] = (
    100 * client_coverage_summary["sku_with_key_customers"]
    / client_coverage_summary["sku_count"]
)

display(client_coverage_summary.round(2))
```

### Итог RFM-анализа

Во втором периоде **Champions и Loyal** составляют **50,80%** известных покупателей и обеспечивают **85,00%** выручки продаж с известным клиентом. Эти сегменты приобретают **91,57% SKU** — почти каждый товар в ассортименте хоть раз купил ценный клиент. Поэтому сам факт покупки ими товара не является основанием для защиты от вывода.

Для оценки риска учитываем **две метрики одновременно**:
- **доля выручки SKU от Champions+Loyal** — насколько товар зависит от ценных клиентов;
- **число ключевых клиентов SKU** — минимум 2, чтобы исключить разовые покупки.

Защита срабатывает только при **совпадении условий**: высокая доля выручки **И** повторные покупки от ключевых. Это отсекает товары, купленные одним ценных клиентом разово, и оставляет те, что действительно держатся на них.

Клиентский анализ охватывает **85,39% общей выручки второго периода**. Остальные **14,61%** — продажи без Customer ID. Отсутствие идентифицированных покупателей нельзя трактовать как низкую клиентскую значимость товара: часть ключевых клиентов могла остаться незамеченной.

В Power BI уникальных покупателей по группам товаров нужно считать через транзакции, а не складывать значения по SKU. Доли выручки пересчитываются из сумм, а не усредняются.

## Углубленный анализ

Цель блока — превратить ABC/XYZ/RFM/BCG-классификацию в обоснованные правила отбора кандидатов на вывод. На выходе — три сценария разной строгости: мягкий, умеренный, агрессивный.

Принцип: сначала формируем защитные признаки, затем на их основе строим сценарии. Ни один сценарий не выводит новые, растущие и сезонные товары без пометки.

### Наблюдаемая история товаров
Что делаем. Определяем первый день продаж каждого SKU и считаем, сколько дней он присутствует в данных на дату оценки периода. Товары с историей ≤ 90 дней не выводим — низкие продажи могут быть следствием недавнего запуска.

```python
sales = clean.loc[~clean["IsCancel"]]
first_sale = sales.groupby("StockCode", observed=True)["InvoiceDate"].min()

assessment_dates = {
    "2009/10": pd.Timestamp("2010-12-01"),
    "2010/11": pd.Timestamp("2011-12-01"),
}

sku_period["first_observed_sale"] = sku_period["StockCode"].map(first_sale)
evaluation_date = pd.to_datetime(
    sku_period["Period"].astype("string").map(assessment_dates)
)

sku_period["observed_history_days"] = (
    evaluation_date - sku_period["first_observed_sale"].dt.normalize()
).dt.days
sku_period["short_history_flag"] = sku_period["observed_history_days"].le(90)

# --- Результат: сводка по флагу короткой истории ---
print("НАБЛЮДАЕМАЯ ИСТОРИЯ ТОВАРОВ")
display(
    sku_period.groupby(["Period", "short_history_flag"], observed=True)
    .agg(
        sku_count=("StockCode", "nunique"),
        revenue=("revenue", "sum"),
        median_history_days=("observed_history_days", "median"),
    )
    .round(2)
)

# --- Конкретно по CZ второго периода ---
cz_p2 = sku_period.loc[
    sku_period["Period"].eq("2010/11")
    & sku_period["ABC_XYZ"].eq("CZ")
]

print("\nCZ второго периода: короткая история")
display(
    cz_p2.groupby("short_history_flag", observed=True)
    .agg(
        sku_count=("StockCode", "size"),
        revenue=("revenue", "sum"),
    )
    .round(2)
)
```

получаем. Флаг short_history_flag для каждого SKU. В CZ второго периода — 139 SKU с короткой историей (выручка 32 331,90 GBP). Их направляем на отдельную оценку, а не в автоматический вывод.

### Повторяющиеся сезонные пики
Проверяем, повторяются ли сезонные пики в оба года. Условие: три крупнейших месяца дают ≥ 50% объёма в каждом году, минимум два пиковых месяца совпадают. Такие товары не выводим — они дают выручку раз в год, но предсказуемо.

```python
# --- Шаг 1. Полные месячные ряды: SKU × календарный месяц (1–12) ---
# Сводим продажи по месяцам для обоих периодов. Пропущенные
# календарные месяцы заполняем нулями, чтобы ряд был полным.
monthly = (
    sku_dynamics.assign(calendar_month=sku_dynamics["Month"].dt.month)
    .pivot_table(
        index=["Period", "StockCode"], columns="calendar_month",
        values="units", aggfunc="sum", observed=True
    )
    .reindex(columns=range(1, 13), fill_value=0)
    .fillna(0)
)

# --- Шаг 2. Разделяем по периодам и берём только сопоставимые SKU ---
# Для проверки сезонности нужны товары, которые продавались в оба года.
p1 = monthly.xs("2009/10")
p2 = monthly.xs("2010/11")
common = p1.index.intersection(p2.index)
p1, p2 = p1.loc[common], p2.loc[common]

print(f"Сопоставимых SKU в обоих периодах: {len(common)}")

# --- Шаг 3. Определяем три крупнейших месяца в каждом периоде ---
def peaks(row):
    """Множество месяцев с тремя крупнейшими объёмами продаж."""
    return set(row[row.gt(0)].nlargest(3).index)

# --- Шаг 4. Доля трёх крупнейших месяцев в общем объёме ---
# Если три месяца дают ≥ 50% — товар имеет выраженный пик.
share1 = p1.apply(lambda x: x.nlargest(3).sum() / x.sum(), axis=1)
share2 = p2.apply(lambda x: x.nlargest(3).sum() / x.sum(), axis=1)

# --- Шаг 5. Пересечение пиковых месяцев между двумя периодами ---
# Минимум 2 совпадающих месяца = сезонный пик повторяется.
overlap = pd.Series(
    [len(a & b) for a, b in zip(p1.apply(peaks, axis=1),
                               p2.apply(peaks, axis=1))],
    index=common
)

# --- Шаг 6. Итоговый флаг: повторяющийся сезонный пик ---
# Три условия одновременно:
#   1) доля топ-3 месяцев ≥ 50% в первом периоде,
#   2) доля топ-3 месяцев ≥ 50% во втором периоде,
#   3) минимум 2 пиковых месяца совпадают.
peak_flag = share1.ge(0.5) & share2.ge(0.5) & overlap.ge(2)

# --- Шаг 7. Переносим флаг в витрину только для второго периода ---
# Для первого периода оставляем пропуск (NA).
# SKU, не сопоставимые с p1 (не продавались в оба периода),
# получают False — сезонность не доказана, блокировать нельзя.
is_p2 = sku_period["Period"].eq("2010/11")
peak_p2 = sku_period["StockCode"].map(peak_flag).fillna(False)
sku_period["repeated_peak_flag"] = (
    peak_p2.where(is_p2).astype("boolean")
)
sku_period["seasonality_checked"] = sku_period["repeated_peak_flag"].notna()

# --- Результат 1. Общая сводка по флагу ---
print("\nПОВТОРЯЮЩИЕСЯ СЕЗОННЫЕ ПИКИ — ПО ПЕРИОДАМ")
display(
    sku_period.groupby(["Period", "repeated_peak_flag"], dropna=False)
    .agg(
        sku_count=("StockCode", "size"),
        revenue=("revenue", "sum"),
    )
    .round(2)
)

# --- Результат 2. Конкретно по CZ второго периода ---
cz_p2 = sku_period.loc[
    sku_period["Period"].eq("2010/11")
    & sku_period["ABC_XYZ"].eq("CZ")
]

print("\nCZ второго периода: повторяющиеся пики")
display(
    cz_p2.groupby("repeated_peak_flag", dropna=False)
    .agg(
        sku_count=("StockCode", "size"),
        revenue=("revenue", "sum"),
    )
    .round(2)
)

# --- Результат 3. Примеры сезонных SKU для ручной проверки ---
seasonal_examples = (
    sku_period.loc[
        sku_period["repeated_peak_flag"].fillna(False)
        & sku_period["Period"].eq("2010/11")
    ]
    .sort_values("revenue", ascending=False)
    [["StockCode", "ProductName", "revenue", "units", "active_months"]]
    .head(10)
)

print("\nТОП-10 сезонных SKU второго периода по выручке")
display(seasonal_examples)
```

Флаг указывает на товары с признаками повторяющейся сезонности — 525 SKU во втором периоде, из них 150 в CZ. Примеры — рождественский декор, грелки — продаются круглый год с выраженным ноябрьско-декабрьским пиком. Это **повод проверить** товар: убедиться, что пик повторится, оценить вклад в новогодний ассортимент и возможность закупки под заказ. Автоматически исключать такие SKU не следует — в согласованных сценариях они исключены из отбора как защита.

### Денежный вклад с учётом отрицательных операций
Считаем сальдо = выручка − возвраты. Товары с неположительным сальдо — приоритетные кандидаты на вывод. Считаем отрицательные операции отдельно, чтобы не терять их при присоединении к витрине.

```python
# --- Шаг 1. Считаем отрицательные операции по каждому SKU × период ---
# Берём только строки с отрицательной суммой: возвраты, отмены,
# переоформления. Меняем знак на положительный, чтобы получить
# абсолютную величину возвратов.
keys = ["Period", "StockCode"]

negative = (
    -clean.loc[clean["Amount"].lt(0)]
    .groupby(keys, observed=True)["Amount"].sum()
).rename("negative_amount")

# --- Шаг 2. Присоединяем к товарной витрине по ключам ---
# join сохраняет все строки sku_period, даже если у SKU нет возвратов.
# Перед повторным запуском удаляем старый столбец, чтобы не было дублей.
sku_period = (
    sku_period.drop(columns="negative_amount", errors="ignore")
    .join(negative, on=keys, validate="one_to_one")
)

# --- Шаг 3. Заполняем пропуски и считаем сальдо ---
# SKU без возвратов: negative_amount = 0.
sku_period["negative_amount"] = sku_period["negative_amount"].fillna(0)

# Сальдо = выручка − возвраты. Это реальный денежный вклад товара.
sku_period["sales_balance"] = (
    sku_period["revenue"] - sku_period["negative_amount"]
)

sku_period["returns_only_flag"] = (
    sku_period["revenue"].eq(0)
    & sku_period["negative_amount"].gt(0)
)

# Доля возвратов в выручке: сигнал для ручной проверки.
sku_period["negative_to_sales_pct"] = (
    100 * sku_period["negative_amount"] / sku_period["revenue"]
)

# --- Результат 1. Общая сводка по периодам ---
print("ДЕНЕЖНЫЙ ВКЛАД С УЧЁТОМ ВОЗВРАТОВ")
display(
    sku_period.groupby("Period", observed=True)
    .agg(
        revenue=("revenue", "sum"),
        negative_amount=("negative_amount", "sum"),
        sales_balance=("sales_balance", "sum"),
    )
    .round(2)
)

# --- Результат 2. Срез по CZ второго периода ---
cz_p2 = sku_period.loc[
    sku_period["Period"].eq("2010/11")
    & sku_period["ABC_XYZ"].eq("CZ")
]

print("\nCZ второго периода: денежный вклад")
display(
    cz_p2[["revenue", "negative_amount", "sales_balance"]]
    .sum()
    .round(2)
    .to_frame()
)

# --- Результат 3. Топ-10 SKU CZ по доле возвратов ---
print("\nТоп-10 CZ по доле возвратов (кандидаты на ручную проверку)")
display(
    cz_p2.sort_values("negative_to_sales_pct", ascending=False)
    [["StockCode", "ProductName", "revenue",
      "negative_amount", "sales_balance", "negative_to_sales_pct"]]
    .head(10)
    .round(2)
)

# --- Результат 4. Топ-10 CZ с отрицательным сальдо ---
print("\nCZ с неположительным сальдо")
display(
    cz_p2.loc[cz_p2["sales_balance"].le(0)]
    [["StockCode", "ProductName", "revenue",
      "negative_amount", "sales_balance"]]
    .sort_values("sales_balance")
    .head(10)
    .round(2)
)
```

Возвраты по магазину растут быстрее выручки (25,5% против 2,5%), но в группе CZ доля возвратов ниже средней — 2,60%. При этом 8 SKU из CZ дают отрицательное или нулевое сальдо — это приоритетные кандидаты для разбора операций. Прежде чем принимать решение о выводе, стоит проверить: связан ли возврат с исходной покупкой, нет ли учётной ошибки или переоформления. Ещё ~10 SKU требуют ручной проверки: возвраты у них превышают продажи — это повод разобраться с качеством, доступностью или учётом.

### Клиентский охват и RFM-значимость
Связываем SKU с сегментами покупателей. Для каждого товара считаем, сколько Champions/Loyal его покупают и какую долю выручки они дают. Это защита от вывода товаров, критичных для ценных клиентов.

```python
# --- Шаг 1. Присоединяем RFM-сегменты к продажам ---
customer_sku_sales = (
    clean.loc[
        ~clean["IsCancel"] & clean["Customer ID"].notna(),
        ["Period", "StockCode", "Customer ID", "Amount"]
    ]
    .merge(
        rfm[["Period", "Customer ID", "RFM_segment"]],
        on=["Period", "Customer ID"],
        how="left",
        validate="many_to_one"
    )
)

# --- Шаг 2. Оставляем только Champions и Loyal ---
key_segments = [
    "Champions — лучшие по RFM",
    "Loyal — повторные покупатели",
]

key_sales = customer_sku_sales.loc[
    customer_sku_sales["RFM_segment"].isin(key_segments)
]

# --- Шаг 3. Считаем две метрики на уровне SKU × период ---
# key_customers             — сколько Champions/Loyal покупали SKU
# key_customer_revenue      — их суммарная выручка (для расчёта доли)
sku_client_metrics = (
    key_sales
    .groupby(["Period", "StockCode"], observed=True)
    .agg(
        key_customers=("Customer ID", "nunique"),
        key_customer_revenue=("Amount", "sum"),
    )
    .reset_index()
)

# --- Шаг 4. Удаляем старые колонки перед merge ---
client_cols = [
    "key_customers", "key_customer_revenue",
    "key_customer_revenue_share_pct",
]

sku_period = (
    sku_period
    .drop(columns=client_cols, errors="ignore")
    .merge(
        sku_client_metrics,
        on=["Period", "StockCode"],
        how="left",
        validate="one_to_one"
    )
)

sku_period["key_customers"] = (
    sku_period["key_customers"].fillna(0).astype("int64")
)
sku_period["key_customer_revenue"] = (
    sku_period["key_customer_revenue"].fillna(0)
)

# --- Шаг 5. Главная метрика риска — доля выручки от Champions/Loyal ---
sku_period["key_customer_revenue_share_pct"] = (
    100 * sku_period["key_customer_revenue"] / sku_period["revenue"]
)

# --- Сводка ---
print("CZ второго периода: зависимость от Champions/Loyal")
cz = sku_period.loc[
    sku_period["Period"].eq("2010/11")
    & sku_period["ABC_XYZ"].eq("CZ")
]

display(
    cz["key_customer_revenue_share_pct"]
    .describe(percentiles=[0.5, 0.75, 0.9, 0.95])
    .round(2)
    .to_frame()
)

display(
    cz.sort_values("key_customer_revenue_share_pct", ascending=False)
    [["StockCode", "ProductName", "revenue",
      "key_customers", "key_customer_revenue_share_pct"]]
    .head(10)
    .round(2)
)
```

Типичный CZ-товар на 60% зависит от Champions/Loyal, поэтому жёсткий фильтр ≤ 40% критичен для сценариев. Но порог нужно дополнить абсолютной суммой (≤ 50 GBP) — иначе мелочь с 100% долей блокирует вывод, хотя реального риска нет.

### Правила отбора и сводка сценариев
Что делаем. Собираем все признаки в единые правила. Формируем три сценария разной строгости на базе CZ. Каждый сценарий сохраняем как булев флаг в витрине.

```python
# --- Шаг 1. Подтягиваем BCG-классификацию по StockCode ---
# BCG_adapted хранится в bcg_common на уровне SKU.
# Переносим в витрину, чтобы использовать признак «растущий».
bcg_map = bcg_common["BCG_adapted"].rename("BCG_adapted")
sku_period["BCG_adapted"] = sku_period["StockCode"].map(bcg_map)

# Отдельное поле для отображения в карточке товара
is_p2 = sku_period["Period"].eq("2010/11")

sku_period["BCG_display"] = sku_period["BCG_adapted"]

sku_period.loc[
    is_p2 & sku_period["returns_only_flag"].fillna(False),
    "BCG_display"
] = "Returns only — продажи прекращены"

sku_period.loc[
    ~is_p2 & sku_period["returns_only_flag"].fillna(False),
    "BCG_display"
] = "Returns only — нет данных о более раннем периоде"

# --- Шаг 2. Базовые булевы признаки без NA ---
# Все признаки приводим к bool или строке — иначе NA ломает фильтры.
is_p2 = sku_period["Period"].eq("2010/11")           # только второй период

# Базовая маска согласована с дашбордом:
#   CZ — товары с низким вкладом и высокой вариативностью;
#   SKU только с возвратами — те, у кого в периоде нет продаж,
#   но есть отрицательные операции. У них нет ABC/XYZ, но они
#   тоже подлежат выводу.
cz = sku_period["ABC_XYZ"].eq("CZ")
returns_only_mask = sku_period["returns_only_flag"]
base = cz | returns_only_mask

new = sku_period["short_history_flag"].fillna(False).astype(bool)
seasonal = sku_period["repeated_peak_flag"].fillna(False).astype(bool)
growing = sku_period["BCG_adapted"].fillna("").isin(
    ["A — рост", "B/C — рост"]
)

# Ядро защиты «без новых, сезонных и растущих» раньше выносилось
# в отдельную переменную safe. Сейчас маска собирается прямо в каждом
# сценарии — переменная не нужна.
# safe = ~new & ~seasonal & ~growing

# --- Шаг 3. Метрики без NA ---
# Покрытие клиентами и доля Champions/Loyal определены только для SKU
# с положительными продажами. Для SKU только с возвратами этих
# показателей нет; приравниваем их к 0, чтобы фильтры сценариев
# не отсеивали такие товары как «неизвестные с нулевым качеством».
no_revenue = sku_period["revenue"].eq(0)

cov = sku_period["customer_revenue_coverage_pct"].fillna(0)
key_share = (
    sku_period["key_customer_revenue_share_pct"]
    .fillna(100)                         # SKU с продажами, но без ключевых
    .where(~no_revenue, 0)               # SKU только с возвратами
)
months = sku_period["active_months"].fillna(0)
balance = sku_period["sales_balance"].fillna(0)

# --- Шаг 4. МЯГКИЙ — безопасный минимум ---
# Согласовано с PBI: ExcludeNew=1, ExcludeSeasonal=1, ExcludeGrowing=1,
# CoverageMin=50, KeyShareMax=30, BalanceMax=1000.
soft = (
    is_p2 & base & ~new & ~seasonal & ~growing
    & cov.ge(50) & key_share.le(30) & balance.le(1000)
)

# --- Шаг 5. УМЕРЕННЫЙ — CZ с ослабленными клиентскими фильтрами ---
# Согласовано с PBI: ExcludeNew=1, ExcludeSeasonal=1, ExcludeGrowing=1,
# CoverageMin=0, KeyShareMax=50, BalanceMax=1000.
moderate = (
    is_p2 & base & ~new & ~seasonal & ~growing
    & cov.ge(0) & key_share.le(50) & balance.le(1000)
)

# --- Шаг 6. АГРЕССИВНЫЙ — цель ~20% ---
# Согласовано с PBI: ExcludeNew=0, ExcludeSeasonal=1, ExcludeGrowing=0,
# CoverageMin=0, KeyShareMax=95, BalanceMax=1000.
# Новые и растущие товары допускаются — защита ослаблена.
aggressive = (
    is_p2 & base & ~seasonal
    & cov.ge(0) & key_share.le(95) & balance.le(1000)
)

# --- Шаг 7. Сохраняем флаги в витрину ---
# where(is_p2) оставляет флаг только для второго периода.
# База — CZ плюс SKU только с возвратами, без захода в CY/BZ/BY.
sku_period["scenario_soft"] = soft.where(is_p2)
sku_period["scenario_moderate"] = moderate.where(is_p2)
sku_period["scenario_aggressive"] = aggressive.where(is_p2)

# --- Шаг 8. Причина ручной проверки для сезонных ---
# Сезонные во всех трёх сценариях исключены — пометка носит справочный
# характер. Она нужна, если бизнес захочет вернуть их в отбор: перед этим
# стоит проверить устойчивость сезонного пика и возможность закупки под заказ.
sku_period["review_reason"] = pd.Series(
    pd.NA, index=sku_period.index, dtype="string"
)
sku_period.loc[seasonal, "review_reason"] = (
    "Сезонный — проверить устойчивость сезонного пика"
)
# --- Результат 1. Сводка по трём сценариям ---
n_total = is_p2.sum()

rows = []
for name, flag in [
    ("Мягкий", "scenario_soft"),
    ("Умеренный", "scenario_moderate"),
    ("Агрессивный", "scenario_aggressive"),
]:
    d = sku_period.loc[sku_period[flag].eq(True)]
    rows.append({
        "Сценарий": name,
        "SKU": len(d),
        "% ассортимента": round(100 * len(d) / n_total, 2),
        "Выручка, GBP": round(d["revenue"].sum(), 2),
        "Сальдо, GBP": round(d["sales_balance"].sum(), 2),
        "Champions/Loyal, GBP": round(d["key_customer_revenue"].sum(), 2),
    })

scenario_summary = pd.DataFrame(rows).set_index("Сценарий")
print("СЦЕНАРИИ ВЫВОДА SKU — СВОДКА")
display(scenario_summary)

# --- Результат 2. Проверка вложенности ---
# Мягкий ⊂ Умеренный ⊂ Агрессивный.
# Все три сценария работают только с CZ — вложенность строгая.
assert (soft & ~moderate).sum() == 0, "Мягкий не вложен в Умеренный"
assert (moderate & ~aggressive).sum() == 0, "Умеренный не вложен в Агрессивный"
print("Проверка вложенности: OK")

# --- Результат 3. Агрессивный vs цель ТЗ ---
target_20 = round(0.20 * n_total)
print(f"\nЦель ТЗ (20%): {target_20} SKU")
print(f"Агрессивный сценарий: {int(aggressive.sum())} SKU "
      f"({100 * aggressive.sum() / n_total:.2f}%)")
```

Агрессивный сценарий даёт 19,84% — практически точно попадает в целевые 20%. Мягкий сценарий — 1,71%, умеренный — 7,29%. Все три сценария вложены строго: мягкий ⊂ умеренный ⊂ агрессивный. Перед сдачей нужно скорректировать фильтры, чтобы шкала была осмысленной, а не просто «три похожих списка».

### Что даёт углублённый анализ

Блок превращает классы ABC/XYZ/RFM/BCG в **обоснованные правила отбора**. Без него проект заканчивается на «вот классы товаров». С ним — на «вот три сценария, вот почему именно они, вот их защита».

| До блока | После блока |
|---|---|
| ABC/XYZ — классы товаров | Кандидаты на вывод |
| RFM — сегменты клиентов | Связь клиентов с товарами |
| BCG — динамика | Защита растущих |
| ETL — признаки | Фильтры отбора |
| Один список | Три сценария с разным риском |

**Что блок не решает:** остатки, маржа, сроки поставки, минимальные партии, обязательства перед покупателями. Это ограничения данных, а не блока.

---

Сценарии определены. Переходим к сборке трёх датасетов.

Углублённый анализ формирует **три сценария вывода SKU**, согласованных с пресетами Power BI:

- **Мягкий** — безопасный минимум: только CZ, без новых, сезонных и растущих; покрытие клиентами ≥ 50%, доля Champions/Loyal ≤ 30%, сальдо ≤ 1000 GBP.
- **Умеренный** — компромисс: только CZ, без новых, сезонных и растущих; покрытие клиентами не ограничено, доля Champions/Loyal ≤ 50%, сальдо ≤ 1000 GBP.
- **Агрессивный** — цель ТЗ (~20%): только CZ, сезонные исключены; новые и растущие товары допускаются, доля Champions/Loyal ≤ 95%, сальдо ≤ 1000 GBP.

**Важно:** в агрессивном сценарии защиты по новизне и росту ослаблены. Его список требует дополнительной проверки на новинки и растущие SKU. Мягкий и умеренный сохраняют эти защиты в полном объёме.

Предусмотрены **три сценария вывода SKU**, вложенных по строгости:

| Сценарий | Целевой % | Логика |
|---|---|---|
| **Мягкий** | ~2% | CZ с жёсткими клиентскими фильтрами — безопасный минимум |
| **Умеренный** | ~5–10% | CZ с ослабленными фильтрами — компромисс |
| **Агрессивный** | ~20% | Весь CZ без новых и растущих + сезонные с пометкой |

Каждый сценарий зафиксирован как булев флаг в витрине `sku_period`. База — CZ плюс SKU только с возвратами. Набор защит различается: мягкий и умеренный исключают новые и растущие товары и ограничивают клиентскую зависимость; агрессивный допускает новые и растущие SKU, но сохраняет ограничение по доле Champions/Loyal и сальдо. Сезонные исключены во всех трёх сценариях.

Превращаем флаги сценариев в **готовый список кандидатов**:

1. Применяем сценарий к витрине и получаем конкретные SKU.
2. Разделяем их по статусам:
   - **Кандидат на вывод** — замечаний нет, можно согласовывать.
   - **Сначала ручная проверка** — сезонные, неположительное сальдо, аномальные возвраты.
3. Для каждого кандидата фиксируем:
   - основание рекомендации;
   - историческую выручку и сальдо;
   - клиентский охват;
   - необходимость ручной проверки.
4. Считаем эффект: сколько SKU и какую долю исторической выручки затрагивает сценарий.
5. Готовим финальную таблицу для согласования с бизнесом.

**Сценарии — не приговор.** Вывод SKU — это решение заказчика, а не автоматическое действие. Историческая выручка кандидатов **не интерпретируется как будущая потеря**: часть спроса может перераспределиться на другие товары, часть — потеряться, часть — остаться без изменений.

**Ограничения данных.** В блоке не учитываются: остатки на складе, маржа и себестоимость, сроки поставки, минимальные партии, обязательства перед покупателями. Историческая выручка кандидатов не интерпретируется как гарантированная будущая потеря. Решение о выводе — за заказчиком, после проверки складских данных. Это ограничения исходных данных, а не блока.

**Отдельно про SKU только с возвратами.** Такие товары входят в базовый набор, потому что формально они не приносят денег. Но прежде чем принимать решение, стоит проверить: не являются ли эти операции корректировками, переоформлениями или разовыми исправлениями. Наличие возврата без соответствующей продажи в периоде может отражать не проблему с товаром, а особенность учёта.

**В витрине такие SKU** получают реальное название из истории операций (продажи или возвраты), флаг `returns_only_flag` и отдельный статус `BCG_display`. Проверка показала: все 14 SKU второго периода продавались в первом — это отложенные возвраты, а не позиции без истории. Признаки не исключают SKU из сценариев и служат контекстом для ручной проверки.


**Три сценария — выбор заказчика.** Мягкий даёт безопасный минимум, агрессивный — цель ТЗ. Бизнес сам выбирает уровень риска, опираясь на сводку и список кандидатов.

```python
# Настройки согласованы с пресетами Power BI (DAX Active Preset).
# Порядок и значения параметров совпадают:
# ExcludeNew, ExcludeSeasonal, ExcludeGrowing, CoverageMin, KeyShareMax, BalanceMax.
#
# Мягкий:     ExcludeNew=1, ExcludeSeasonal=1, ExcludeGrowing=1,
#             CoverageMin=50, KeyShareMax=30,  BalanceMax=1000
# Умеренный:  ExcludeNew=1, ExcludeSeasonal=1, ExcludeGrowing=1,
#             CoverageMin=0,  KeyShareMax=50,  BalanceMax=1000
# Агрессивный: ExcludeNew=0, ExcludeSeasonal=1, ExcludeGrowing=0,
#             CoverageMin=0,  KeyShareMax=95,  BalanceMax=1000
#
# Важно: агрессивный сценарий допускает новые и растущие товары.
# Это осознанно ослабленная защита — список требует дополнительной
# проверки на наличие новинок и растущих SKU.
SCENARIOS = {
    "Мягкий": {
        "groups": ["CZ"],
        "exclude_new": True,           # ExcludeNew = 1
        "exclude_seasonal": True,      # ExcludeSeasonal = 1
        "exclude_growing": True,       # ExcludeGrowing = 1
        "cov_min": 50,                 # CoverageMin = 50
        "key_share_max": 30,           # KeyShareMax = 30
        "balance_max": 1000,           # BalanceMax = 1000
        "months_max": 12,              # фильтр по месяцам в PBI не используется
        "target_pct": 2,
    },
    "Умеренный": {
        "groups": ["CZ"],
        "exclude_new": True,           # ExcludeNew = 1
        "exclude_seasonal": True,      # ExcludeSeasonal = 1
        "exclude_growing": True,       # ExcludeGrowing = 1
        "cov_min": 0,                  # CoverageMin = 0
        "key_share_max": 50,           # KeyShareMax = 50
        "balance_max": 1000,           # BalanceMax = 1000
        "months_max": 12,
        "target_pct": 5,
    },
    "Агрессивный": {
        "groups": ["CZ"],
        "exclude_new": False,          # ExcludeNew = 0 (новые допускаются)
        "exclude_seasonal": True,      # ExcludeSeasonal = 1
        "exclude_growing": False,      # ExcludeGrowing = 0 (растущие допускаются)
        "cov_min": 0,                  # CoverageMin = 0
        "key_share_max": 95,           # KeyShareMax = 95
        "balance_max": 1000,           # BalanceMax = 1000
        "months_max": 12,
        "target_pct": 20,
    },
}
```

```python
# ============================================================
# ПРИМЕНЕНИЕ СЦЕНАРИЕВ
# ============================================================

# Базовые признаки
is_p2 = sku_period["Period"].eq("2010/11")
new = sku_period["short_history_flag"].fillna(False).astype(bool)
seasonal = sku_period["repeated_peak_flag"].fillna(False).astype(bool)
growing = sku_period["BCG_adapted"].fillna("").isin(["A — рост", "B/C — рост"])

# Метрики
# Для SKU только с возвратами (revenue = 0) метрики, зависящие от выручки,
# неопределены. Приводим их к значениям, которые НЕ блокируют фильтры —
# та же логика, что и в блоке «Правила отбора и сводка сценариев».
no_revenue = sku_period["revenue"].eq(0)

cov = (
    sku_period["customer_revenue_coverage_pct"]
    .fillna(0)
    .where(~no_revenue, 100)      # returns-only: покрытие не блокирует
)
key_share = (
    sku_period["key_customer_revenue_share_pct"]
    .fillna(100)
    .where(~no_revenue, 0)         # returns-only: нет ключевых клиентов
)
balance = sku_period["sales_balance"].fillna(0)
months = (
    sku_period["active_months"]
    .fillna(0)
    .where(~no_revenue, 12)        # returns-only: активны, не отсеиваются
)

# Группы
group_mask = {
    g: sku_period["ABC_XYZ"].eq(g) for g in ["AX", "AY", "AZ", "BX", "BY", "BZ", "CX", "CY", "CZ"]
}

n_total = is_p2.sum()

# Базовая маска: CZ + SKU только с возвратами.
# Согласовано с PBI: базовый набор шире, чем просто CZ.
returns_only_mask = sku_period["returns_only_flag"]

# ---- Функция применения одного сценария ----
def apply_scenario(cfg):
    """Собрать булеву маску по конфигу сценария."""
    mask = is_p2 & group_mask[cfg["groups"][0]]
    for g in cfg["groups"][1:]:
        mask = mask | (is_p2 & group_mask[g])
    # Добавляем SKU только с возвратами — они входят в базовый набор.
    mask = mask | (is_p2 & returns_only_mask)

    if cfg["exclude_new"]:
        mask &= ~new
    if cfg["exclude_seasonal"]:
        mask &= ~seasonal
    if cfg["exclude_growing"]:
        mask &= ~growing

    mask &= cov.ge(cfg["cov_min"])
    mask &= key_share.le(cfg["key_share_max"])
    mask &= balance.le(cfg["balance_max"])
    mask &= months.le(cfg["months_max"])

    return mask

# ---- Применяем все сценарии ----
rows = []
masks = {}
for name, cfg in SCENARIOS.items():
    mask = apply_scenario(cfg)
    masks[name] = mask

    d = sku_period.loc[mask]
    pct = 100 * len(d) / n_total

    rows.append({
        "Сценарий": name,
        "SKU": len(d),
        "% ассортимента": round(pct, 2),
        "Цель, %": cfg["target_pct"],
        "Отклонение, п.п.": round(pct - cfg["target_pct"], 2),
        "Выручка, GBP": round(d["revenue"].sum(), 2),
        "Сальдо, GBP": round(d["sales_balance"].sum(), 2),
        "Champions/Loyal, GBP": round(d["key_customer_revenue"].sum(), 2),
    })

summary = pd.DataFrame(rows).set_index("Сценарий")
print("СЦЕНАРИИ ВЫВОДА SKU")
display(summary)

# ---- Проверка вложенности ----
print("\nВложенность (мягкий ⊂ умеренный ⊂ агрессивный):")
soft = masks["Мягкий"]
moderate = masks["Умеренный"]
aggressive = masks["Агрессивный"]

print(f"  Мягкий ⊂ Умеренный:      {(soft & ~moderate).sum()} нарушений")
print(f"  Умеренный ⊂ Агрессивный: {(moderate & ~aggressive).sum()} нарушений")

# ---- Пометки для ручной проверки в агрессивном ----
seasonal_in_aggressive = aggressive & seasonal
print(f"\nСезонных в агрессивном (на ручную проверку): {seasonal_in_aggressive.sum()}")
```

Финальные настройки определены. Отклонения от целевых значений минимальны:

Переходим к оформлению окончательных списков.

Применяем три зафиксированных сценария к витрине, разделяем кандидатов по статусам, считаем эффект.

```python
# ============================================================
# ТРИ ДАТАСЕТА КАНДИДАТОВ ПО СЦЕНАРИЯМ
# ============================================================

# --- Базовые признаки ---
is_p2 = sku_period["Period"].eq("2010/11")
new = sku_period["short_history_flag"].fillna(False).astype(bool)     # новый (<90 дней)
seasonal = sku_period["repeated_peak_flag"].fillna(False).astype(bool)  # сезонный
growing = sku_period["BCG_adapted"].fillna("").isin(["A — рост", "B/C — рост"])

# --- Метрики для фильтров ---
# Согласовано с блоком «ПРИМЕНЕНИЕ СЦЕНАРИЕВ» и PBI:
# для returns-only метрики заданы так, чтобы фильтры их не блокировали.
no_revenue = sku_period["revenue"].eq(0)

cov = (
    sku_period["customer_revenue_coverage_pct"]
    .fillna(0)
    .where(~no_revenue, 100)
)
key_share = (
    sku_period["key_customer_revenue_share_pct"]
    .fillna(100)
    .where(~no_revenue, 0)
)
balance = sku_period["sales_balance"].fillna(0)
months = (
    sku_period["active_months"]
    .fillna(0)
    .where(~no_revenue, 12)
)

n_total = is_p2.sum()

# --- Базовая маска: CZ + SKU только с возвратами ---
returns_only_mask = sku_period["returns_only_flag"]

# --- Функция: маска одного сценария по конфигу ---
def apply_scenario(cfg):
    mask = is_p2 & sku_period["ABC_XYZ"].isin(cfg["groups"])
    # SKU только с возвратами входят в базовый набор во всех сценариях.
    mask = mask | (is_p2 & returns_only_mask)
    if cfg["exclude_new"]:       mask &= ~new
    if cfg["exclude_seasonal"]:  mask &= ~seasonal
    if cfg["exclude_growing"]:   mask &= ~growing
    mask &= cov.ge(cfg["cov_min"])
    mask &= key_share.le(cfg["key_share_max"])
    mask &= balance.le(cfg["balance_max"])
    mask &= months.le(cfg["months_max"])
    return mask

# --- Признаки ручной проверки ---
manual = (
    seasonal
    | balance.le(0)
    | sku_period["negative_to_sales_pct"].fillna(0).gt(50)
    | key_share.gt(80)
)

reason = pd.Series(pd.NA, index=sku_period.index, dtype="string")
reason.loc[balance.le(0)] = "Неположительное сальдо"
reason.loc[seasonal & reason.isna()] = "Сезонный — переводить на заказ"
reason.loc[sku_period["negative_to_sales_pct"].fillna(0).gt(50) & reason.isna()] = "Аномальные возвраты"
reason.loc[key_share.gt(80) & reason.isna()] = "Зависимость от Champions/Loyal"

# --- Колонки датасета ---
cols = [
    "Period", "StockCode", "ProductName", "ABC_XYZ",
    "revenue", "negative_amount", "sales_balance",
    "key_customers", "key_customer_revenue",
    "key_customer_revenue_share_pct", "customer_revenue_coverage_pct",
    "active_months", "orders",
]

# --- Собираем три датасета ---
datasets = {}
for name, cfg in SCENARIOS.items():
    mask = apply_scenario(cfg)
    datasets[name] = (
        sku_period.loc[mask, cols]
        .assign(
            Status=np.where(manual.loc[mask],
                            "Сначала ручная проверка",
                            "Кандидат на вывод — согласовать"),
            Reason=reason.loc[mask],
        )
        .sort_values(["Status", "revenue"])
        .reset_index(drop=True)
    )

# --- Сводка ---
rows = []
for name, ds in datasets.items():
    cfg = SCENARIOS[name]
    status_counts = ds.groupby("Status", observed=True)["StockCode"].nunique()
    rows.append({
        "Сценарий": name,
        "SKU": len(ds),
        "К выводу": int(status_counts.get("Кандидат на вывод — согласовать", 0)),
        "Ручная проверка": int(status_counts.get("Сначала ручная проверка", 0)),
        "% ассортимента": round(100 * len(ds) / n_total, 2),
        "Цель, %": cfg["target_pct"],
        "Выручка, GBP": round(ds["revenue"].sum(), 2),
        "Сальдо, GBP": round(ds["sales_balance"].sum(), 2),
        "Champions/Loyal, GBP": round(ds["key_customer_revenue"].sum(), 2),
    })
print("СЦЕНАРИИ ВЫВОДА SKU")
display(pd.DataFrame(rows).set_index("Сценарий"))

# --- Вложенность ---
soft = set(datasets["Мягкий"]["StockCode"])
mod = set(datasets["Умеренный"]["StockCode"])
agg = set(datasets["Агрессивный"]["StockCode"])
print(f"\nВложенность: мягкий→умеренный {len(soft - mod)} нарушений, "
      f"умеренный→агрессивный {len(mod - agg)} нарушений")

# --- Готовые датасеты ---
candidates_soft = datasets["Мягкий"]
candidates_moderate = datasets["Умеренный"]
candidates_aggressive = datasets["Агрессивный"]

# --- Сезонные в агрессивном ---
n_seasonal = (candidates_aggressive["Reason"] == "Сезонный — переводить на заказ").sum()
print(f"Сезонных в агрессивном на ручную проверку: {n_seasonal}")

display(candidates_aggressive.head(10).round(2))
```

**Созданы три датасета:** `candidates_soft`, `candidates_moderate`, `candidates_aggressive`. Каждый содержит код товара, название, класс ABC/XYZ, выручку, сальдо, клиентский охват, статус кандидата и причину ручной проверки.

|---|---|---|---|---|---|

Вложенность сохранена: мягкий ⊂ умеренный ⊂ агрессивный.

Внутри каждого списка — два статуса

**Кандидаты на вывод — согласовать:**
SKU без защитных признаков. Их можно рассматривать к выводу после согласования с бизнесом и проверки складских данных.

**Сначала ручная проверка:**
- 0 сезонных в самом списке агрессивного — фильтр их исключает, но по ~525 SKU из p2 нужно отдельно решить: заказная модель
- SKU с неположительным сальдо — не приносят денег
- SKU с аномальными возвратами (>50% от выручки) — разобраться с операциями
- SKU с высокой зависимостью от Champions/Loyal (>80%) — риск потери клиентов

Что это значит

**Главный вывод:** агрессивный сценарий охватывает 19,84% ассортимента, но только **~1,2% исторической выручки**. То есть 756 SKU из 3 797 приносят 118 771 GBP из 9,6 млн.

**Сезонные товары требуют отдельного решения.** Флаг повторяющихся пиков указывает на признаки устойчивой сезонности — это повод проверить товар, а не автоматически переводить его на заказную модель. Перед решением стоит убедиться, что пик повторяется и в следующем году, оценить вклад в новогодний ассортимент и возможность закупки под заказ. В согласованных сценариях такие SKU исключены из списка кандидатов — это защита.

**Защиты различаются по сценариям.** В мягком и умеренном новые и растущие товары исключены, клиентская зависимость ограничена жёстко. В агрессивном сценарии защиты по новизне и росту ослаблены — такие SKU попадают в список и требуют дополнительной проверки. Клиентская зависимость во всех сценариях ограничена порогом KeyShareMax.

**Историческая выручка — не прогноз потерь.** Выручка кандидатов за прошлый год — не оценка будущих потерь. После вывода часть спроса может перераспределиться на другие товары, часть — потеряться, часть — остаться без изменений. Оценить реальный эффект без данных о марже, остатках и условиях поставок нельзя.

Рекомендации заказчику

1. Начать с агрессивного сценария — как ориентир по объёму
Он ближе всего к целевым 20%. Но в нём ослаблены защиты по новизне и росту, поэтому перед согласованием список стоит проверить на наличие новинок и растущих SKU — они попадают в отбор и требуют отдельного решения. Мягкий и умеренный — альтернативы, если бизнес предпочтёт более осторожный подход.

2. Разделить список на две очереди

**Первая очередь — к выводу:**
- SKU без защитных признаков
- С неположительным сальдо — приоритет для разбора операций: нужно понять, связан ли возврат с исходной покупкой или это отдельная корректировка
- SKU только с возвратами (без продаж в периоде) — такие товары в базовом наборе сценариев и требуют решения по сути
- С минимальной клиентской зависимостью

**Вторая очередь — на проверку:**
- ~525 сезонных SKU из p2 — они не входят в отбор ни в одном сценарии; отдельное решение — заказная модель
- С аномальными возвратами — разобраться с качеством/учётом
- С высокой зависимостью от Champions/Loyal — проверить вручную

3. Проверить перед выводом

| Что проверить | Зачем |
|---|---|
| **Остатки на складе** | Товар может лежать мёртвым грузом — тогда вывод ускорит освобождение |
| **Себестоимость и маржа** | Товар с низкой выручкой может быть высокомаржинальным |
| **Сроки поставки** | Долгий lead time = нельзя вывести без замены |
| **Минимальные партии** | Поставщик может требовать закупку партией |
| **Обязательства перед покупателями** | Клиенты могут ждать конкретный товар |

4. Мониторить после вывода

- Изменение общей выручки через 3–6 месяцев
- Перераспределение спроса на другие SKU
- Потерю клиентов, которые покупали только выводимые товары
- Появление новых товаров, которые могут заменить выведенные
